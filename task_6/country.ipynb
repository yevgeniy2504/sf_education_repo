```python
unique_cuisines = []

for recipe in recipes:
    cuisine = recipe.get("cuisine")
    if cuisine and cuisine not in unique_cuisines:
        unique_cuisines.append(cuisine)

print(f"Количество уникальных кухонь: {len(unique_cuisines)}")
print("Список кухонь:", unique_cuisines)
```

```python
# Словарь для хранения количества рецептов на каждую кухню
cuisine_counts = {}

# Проходим по каждой уникальной кухне
for cuisine in unique_cuisines:
    count = 0
    for recipe in recipes:
        if recipe.get("cuisine") == cuisine:
            count += 1
    cuisine_counts[cuisine] = count

# Находим кухню с максимальным числом рецептов
most_common_cuisine = max(cuisine_counts, key=cuisine_counts.get)
max_count = cuisine_counts[most_common_cuisine]

print(f"Больше всего рецептов принадлежит кухне: {most_common_cuisine}")
print(f"Количество рецептов: {max_count}")
```

```python
# Импортируем модуль pandas
import pandas as pd 
# Создаём объект DataFrame, загружая содержимое файла recipes.json
df = pd.read_json('data/recipes.json') 

# Выводим на экран первые строки полученного DataFrame
display(df.head())
```

```python
ids = df['id'].drop_duplicates().tolist()

# Создаем список ингредиентов — все колонки, кроме служебных
ingredients = [col for col in df.columns if col not in ['cuisine', 'id', 'ingredients']]

def make_list(row):
    # Выбираем только те ингредиенты, у которых значение в этой строке != 0
    present_ingredients = [ingredient for ingredient in ingredients if row[ingredient] != 0]
    return present_ingredients


# Создаём пустой список для хранения итоговой структуры
new_recipes = [] 
# Организуем цикл с параметром current_id
for current_id in ids: 
    cuisine = df[df['id'] == current_id]['cuisine'].iloc[0] 
    current_ingredients = make_list(df[df['id'] == current_id].iloc[0]) 
    current_recipe = {'cuisine': cuisine, 'id': int(current_id), 'ingredients': current_ingredients} 
    new_recipes.append(current_recipe)


# Импорт модуля json
import json 
# Функция dumps() модуля json сериализирует объект Python в строку формата JSON. 
new_recipes = json.dumps(new_recipes) 

# Откроем файл new_recipes.json для записи
with open("data/new_recipes.json", "w") as write_file: 
    # Записываем содержимое подготовленные данные в файл
    write_file.write(new_recipes)
```

```python
import requests
from bs4 import BeautifulSoup

def wiki_header(url):
    response = requests.get(url, verify = False)
    soup = BeautifulSoup(response.content, 'html.parser')
    header = soup.find('h1')
    return header.text.strip() if header else None
```
