```python
import requests
from pprint import pprint
import time

token = '62164e2762164e2762164e2736612390c86621662164e270a7cdb0effff12e04978e5de'
url = 'https://api.vk.com/method/users.get'
version = '5.199'

female = 0
known_gender = 0

for i in range(1, 501, 100):  # шаг по 100 ID
    ids = ",".join(map(str, range(i, i + 100)))
    params = {
        'user_ids': ids,
        'access_token': token,
        'v': version,
        'fields': 'sex'
    }

    response = requests.get(url, params=params)
    data = response.json()

    for user in data.get('response', []):
        sex = user.get('sex', 0)
        if sex == 1:
            female += 1
            known_gender += 1
        elif sex == 2:
            known_gender += 1
        # sex == 0 пропускаем

    time.sleep(0.34)  # Не более 3 запросов в секунду

# Вычисляем долю
if known_gender > 0:
    share = round(female / known_gender, 2)
    print(share)
else:
    print("Нет пользователей с определённым полом.")
```
