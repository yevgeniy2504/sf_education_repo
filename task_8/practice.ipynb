```python
from scipy import stats

norm = 75

data = [77, 78, 74, 81, 87, 73, 80, 72, 73, 81, 88, 74, 74, 79, 78, 73, 89, 82, 77, 77, 88, 75, 80, 88, 72, 84, 88, 74, 74, 78]

result_two_sided = stats.ttest_1samp(data, popmean=norm, alternative='two-sided')
print("3) mean != norm :", result_two_sided)
```

```python
from statsmodels.stats.descriptivestats import sign_test


stat6, p6 = sign_test(data, mu0=75)
print(stat6, p6)
```

```python
import numpy as np
from scipy.stats import kruskal


# ================================================================
# 📘 ФУНКЦИЯ: Kruskal–Wallis H-test
# ================================================================
def kruskal_test(*groups):
    """
    Выполняет Kruskal–Wallis тест для k независимых групп.

    Параметры:
        groups — произвольное число массивов данных

    Возвращает:
        H-statistic, p-value
    """
    return kruskal(*groups)

# «Тарифы»:


data_tariffs = [57, 56, 64, 62, 64, 63, 57, 64, 57, 63, 62, 59, 61, 61, 62, 57, 55, 60, 56, 62, 64, 56, 62, 62, 65]

# «Спецпредложения»:

data_offers = [56, 56, 60, 64, 62, 63, 63, 56, 64, 60, 58, 61, 59, 63, 62, 63, 59, 64, 63, 58, 61, 61, 62, 55, 61]

# «Техподдержка»:

data_support = [65, 67, 65, 69, 66, 65, 62, 63, 65, 69, 64, 61, 69, 69, 62, 68, 65, 69, 62, 67, 63, 69, 63, 67, 61]


result = kruskal_test(data_tariffs, data_offers, data_support)

print(result)

```
