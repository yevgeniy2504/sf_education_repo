```python
import pandas as pd
import numpy as np

data = pd.read_csv('data/model.csv')
```

```python
#  В случае большого количества данных в датасете матрица, возвращаемая методом pandas data.corr(), становится нечитаемой.
# Задание 6.1
# 1 point possible (graded)
# Постройте матрицу корреляций для датасета.
# В ответ впишите самый высокий коэффициент корреляции. Ответ округлите до сотых.
```

```python
corr = data.corr(numeric_only=True)
max_corr = corr.where(~np.eye(len(corr), dtype=bool)).max().max()
print(round(max_corr, 2))
```

```python
import seaborn as sns
sns.heatmap(data.corr(numeric_only=True), annot=True)
```

```python
# Посчитайте коэффициент корреляции между признаками Weight (вес модели)
# и Height (рост модели). Ответ округлите до сотых.
corr_weight_height = data['Weight'].corr(data['Height'])
print(round(corr_weight_height, 2))
```

```python
# Рассчитайте среднее значение признаков при помощи библиотеки statistics для признаков weight и height. Ответ округлите до сотых.

import statistics as stats
mean_weight = round(stats.mean(data['Weight']), 2)
mean_height = round(stats.mean(data['Height']), 2)
print(mean_weight, mean_height)
```
