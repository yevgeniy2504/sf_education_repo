# 🔵 1. ОДНА ГРУППА (сравнение с числом)

## 1.1 One-sample t-test  
**Когда:** одна выборка, нормальность, сравнение со средним значением  
**Пример:** проверить, стало ли среднее время доставки меньше 40 минут  
(одна группа, сравнение с нормативом)

### 1.1.a Bootstrap mean test  
**Когда:** нет нормальности, выбросы  
**Пример:** оценить средний расход топлива автомобиля при "грязных" данных

### 1.1.b Permutation one-sample test  
**Когда:** вообще не хотим предположений  
**Пример:** проверить, превышает ли средняя зарплата целевой показатель

### 1.1.c One-sample Wilcoxon signed-rank  
**Когда:** ненормальность + симметрия  
**Пример:** сравнить медианное время реакции оператора с нормативом

---

## 1.2 Sign-test  
**Когда:** одна выборка, ненормальные данные, проверка медианы  
**Пример:** проверить, изменилась ли медиана количества ошибок сотрудника

### 1.2.a One-sample Wilcoxon  
**Когда:** мощнее Sign-test  
**Пример:** медианное время обработки заявки относительно стандарта

### 1.2.b Median test (1-sample)  
**Когда:** сильные выбросы  
**Пример:** проверить, превышает ли медиана температуры оборудования норму

---

# 🟢 2. ДВЕ НЕЗАВИСИМЫЕ ГРУППЫ

## 2.1 Independent t-test  
**Когда:** 2 группы, нормальность, равные дисперсии  
**Пример:** сравнить эффективность двух лекарств (среднее снижение симптомов)

### 2.1.a Welch t-test  
**Когда:** разные дисперсии  
**Пример:** сравнить выработку двух заводов, когда вариабельность разная

### 2.1.b Permutation 2-sample test  
**Когда:** нет нормальности  
**Пример:** сравнить среднюю прибыль двух магазинов без предположений

### 2.1.c Kolmogorov–Smirnov 2-sample  
**Когда:** сравнение форм распределений  
**Пример:** проверить, отличаются ли профили продаж в двух городах вообще

### 2.1.d Brunner–Munzel test  
**Когда:** устойчиво при разных распределениях  
**Пример:** сравнить время отклика двух серверов при разной волатильности

---

## 2.2 Mann–Whitney U-test  
**Когда:** 2 группы, ненормальные данные  
**Пример:** сравнить оценки качества обслуживания в двух филиалах

### 2.2.a Brunner–Munzel  
**Когда:** форма распределений разная  
**Пример:** сравнить время загрузки сайта в 2 странах

### 2.2.b Kolmogorov–Smirnov  
**Когда:** сравнение формы  
**Пример:** сравнить распределение доходов жителей 2 городов

### 2.2.c Median test (2-sample)  
**Когда:** сравнение медиан  
**Пример:** сравнить медианные сроки доставки у двух служб

---

# 🟣 3. ДВЕ ЗАВИСИМЫЕ ГРУППЫ (до/после)

## 3.1 Paired t-test  
**Когда:** зависимые, нормальные  
**Пример:** проверить, увеличилась ли производительность сотрудника после обучения

### 3.1.a Permutation paired test  
**Когда:** нет нормальности  
**Пример:** сравнить время выполнения задачи до/после оптимизации процесса

### 3.1.b Bayesian paired test  
**Когда:** хотим вероятность улучшения  
**Пример:** "с вероятностью 90% производительность выросла?"

---

## 3.2 Wilcoxon signed-rank  
**Когда:** зависимые, ненормальные  
**Пример:** проверить, снизилось ли число ошибок после внедрения нового интерфейса

### 3.2.a Sign-test  
**Когда:** разности асимметричны  
**Пример:** количество дефектов "до/после" при нерегулярных изменениях

### 3.2.b Paired bootstrap  
**Когда:** много выбросов  
**Пример:** время обработки звонков "до/после" при сильных колебаниях

---

# 🟠 3+ НЕЗАВИСИМЫХ ГРУПП

## 4.1 One-way ANOVA  
**Когда:** 3+ групп, нормальные данные  
**Пример:** сравнить среднее время доставки в трёх районах города

### 4.1.a Welch ANOVA  
**Когда:** разные дисперсии  
**Пример:** сравнить зарплаты в 3 компаниях с разной вариабельностью

### 4.1.b Permutation ANOVA  
**Когда:** не хотим нормальности  
**Пример:** сравнить средний доход 3 подразделений без предположений

### 4.1.c Brown–Forsythe ANOVA  
**Когда:** выбросы + разные дисперсии  
**Пример:** сравнить выработку 3 производственных линий с нестабильностью

---

## 4.2 Kruskal–Wallis  
**Когда:** 3+ групп, ненормальные  
**Пример:** сравнить оценки клиентов в трёх филиалах

### 4.2.a Median test (k-sample)  
**Когда:** сравнение медиан  
**Пример:** медианные продажи в трёх магазинах

### 4.2.b Jonckheere–Terpstra  
**Когда:** группы имеют порядок  
**Пример:** проверка, растёт ли удовлетворенность от 1-го до 3-го филиала

### 4.2.c Permutation k-sample  
**Когда:** полная гибкость  
**Пример:** сравнить среднюю прибыль трёх складов без предположений

---

# 🟡 3+ ЗАВИСИМЫХ ИЗМЕРЕНИЙ (повторные)

## 5.1 Repeated Measures ANOVA  
**Когда:** 3+ зависимых измерений, нормальность  
**Пример:** оценить скорость работы утром/днём/вечером у тех же сотрудников

### 5.1.a Linear Mixed Models  
**Когда:** пропуски, разное расстояние между измерениями  
**Пример:** многомесячное наблюдение сотрудников, где есть пропуски

### 5.1.b MANOVA  
**Когда:** несколько зависимых переменных  
**Пример:** сравнить изменения времени + качества работы по трём сессиям

---

## 5.2 Friedman test  
**Когда:** 3+ зависимых, ненормальные  
**Пример:** сравнить производительность на 3 версиях интерфейса у тех же людей

### 5.2.a Quade test  
**Когда:** различия между уровнями должны учитываться  
**Пример:** оценка улучшений при тренировке на 3 типах упражнений

### 5.2.b Aligned Rank Transform ANOVA  
**Когда:** ранги + нужен факторный дизайн  
**Пример:** изучение влияния времени дня и интерфейса одновременно

```python
# =====================================================================
# 📘 ONE-SAMPLE T-TEST — ВСЕ СТАНДАРТНЫЕ ВАРИАНТЫ ЗАДАЧ
# =====================================================================
# ДАННЫЕ: одна группа чисел
data = [31, 38, 30, 46, 40, 36, 38, 44]

# НОРМАТИВ / ЦЕЛЕВОЕ ЧИСЛО
norm = 40


# =====================================================================
# 1) ПРОВЕРИТЬ: СРЕДНЕЕ МЕНЬШЕ НОРМЫ  (mean < norm)
# =====================================================================
# Пример задачи:
# • время доставки < 40?
# • температура оборудования ниже нормы?
# • средние продажи упали?
#
# H0: mean = norm
# H1: mean < norm
result_less = stats.ttest_1samp(data, popmean=norm, alternative='less')
print("1) mean < norm :", result_less)


# =====================================================================
# 2) ПРОВЕРИТЬ: СРЕДНЕЕ БОЛЬШЕ НОРМЫ  (mean > norm)
# =====================================================================
# Пример задачи:
# • производительность выросла > стандарта?
# • средняя зарплата выше 300 000?
#
# H0: mean = norm
# H1: mean > norm
result_greater = stats.ttest_1samp(data, popmean=norm, alternative='greater')
print("2) mean > norm :", result_greater)


# =====================================================================
# 3) ПРОВЕРИТЬ: СРЕДНЕЕ НЕ РАВНО НОРМЕ  (mean != norm)
# =====================================================================
# Пример задачи:
# • потребление энергии изменилось? (неважно, больше или меньше)
# • средняя длительность звонка изменилась?
#
# H0: mean = norm
# H1: mean != norm
result_two_sided = stats.ttest_1samp(data, popmean=norm, alternative='two-sided')
print("3) mean != norm :", result_two_sided)


# =====================================================================
# 4) ПРОВЕРИТЬ: СРЕДНЕЕ РАВНО НОРМЕ  (mean = norm)
# =====================================================================
# ВАЖНО:
# В t-test проверка "равно" = это всегда H0.
# Специального вызова НЕТ.
#
# Интерпретация:
# • если p-value большое → НЕТ оснований отвергнуть mean = norm
# • если p-value маленькое → среднее ≠ norm
#
print("4) Проверка равенства — это интерпретация результата теста выше")


# =====================================================================
# 5) ПРОВЕРИТЬ: МЕДИАНУ (если данные НЕ нормальные)
# =====================================================================
# Для медианы используется НЕ t-test:
#   • Sign-test
#   • Wilcoxon signed-rank (если симметрия)
#
# Проверить: median < norm
from statsmodels.stats.descriptivestats import sign_test

result_median = sign_test(data, mu0=norm)   # всегда двусторонний тест
print("5) median != norm (Sign-test):", result_median)

# Примеры задач:
# • медианное время реакции ≠ стандарту
# • медианное время обработки заявок изменилось
# • медианная зарплата ≠ целевому уровню


# =====================================================================
# 6) ПРИМЕРЫ РЕАЛЬНЫХ ЗАДАЧ ДЛЯ ONE-SAMPLE T-TEST
# =====================================================================
# • Время доставки < нормы?
# • Средняя температура > допустимого порога?
# • Средняя производительность выросла > стандарта?
# • Средняя длительность звонка ≠ прошлому месяцу?
# • Средний срок ремонта = регламенту?
# • Средняя зарплата > установленного минимума?
# • Средний расход топлива < паспортного значения?
#
# Для всех этих задач используется t-test, если данные нормальные
# и альтернатива ВСЕГДА задается через:
#   alternative='less' / 'greater' / 'two-sided'
#
# =====================================================================
```

```python
# ================================================================
# 📘 1.1.a ФУНКЦИЯ: Bootstrap Mean Test
# ================================================================
# Bootstrap-тест на среднее применяется когда:
#   • данные НЕ нормальные
#   • есть выбросы
#   • малый размер выборки (n < 30)
#   • нужно тестировать ИМЕННО СРЕДНЕЕ, а не медиану
#
# Проверяем:
#   H0: mean(data) = m0
#   H1: mean(data) < m0   (alternative='less')
#   H1: mean(data) > m0   (alternative='greater')
#   H1: mean(data) != m0  (alternative='two-sided')
#
# ================================================================

import numpy as np


def bootstrap_mean_test(data, m0, alternative='two-sided', n_boot=20000):
    """
    Bootstrap проверка среднего.
    data — список или numpy массив
    m0 — число, с которым сравниваем среднее
    alternative — 'less', 'greater', 'two-sided'
    """
    data = np.array(data)
    n = len(data)

    # Bootstrap выборки
    boot_means = np.array([
        np.mean(np.random.choice(data, size=n, replace=True))
        for _ in range(n_boot)
    ])

    # Проверка гипотез
    if alternative == 'less':
        p = np.mean(boot_means < m0)

    elif alternative == 'greater':
        p = np.mean(boot_means > m0)

    elif alternative == 'two-sided':
        diff = abs(np.mean(data) - m0)
        p = np.mean(abs(boot_means - m0) >= diff)

    else:
        raise ValueError("alternative must be 'less', 'greater', or 'two-sided'")

    return p, np.mean(data), boot_means


# ================================================================
# 📘 1.1.a.1 Пример — mean < m0
# ================================================================
# Задача: среднее время доставки МЕНЬШЕ 40?
data1 = [31, 38, 30, 46, 40, 36, 38, 44]
p1, mean1, _ = bootstrap_mean_test(data1, m0=40, alternative='less')
print("1.1.a.1 mean < 40 :", p1, "| mean =", mean1)


# ================================================================
# 📘 1.1.a.2 Пример — mean > m0
# ================================================================
# Задача: производительность БОЛЬШЕ 100?
data2 = [102, 99, 110, 108, 112, 115, 98]
p2, mean2, _ = bootstrap_mean_test(data2, m0=100, alternative='greater')
print("1.1.a.2 mean > 100 :", p2, "| mean =", mean2)


# ================================================================
# 📘 1.1.a.3 Пример — mean != m0 (двусторонний)
# ================================================================
# Задача: изменилось ли среднее потребление энергии ≠ 500?
data3 = [480, 520, 505, 499, 630, 520, 480, 700]
p3, mean3, _ = bootstrap_mean_test(data3, m0=500, alternative='two-sided')
print("1.1.a.3 mean != 500 :", p3, "| mean =", mean3)


# ================================================================
# 📘 1.1.a.4 Пример — ненормальные данные (экспонента)
# ================================================================
data4 = np.random.exponential(scale=10, size=50)
p4, mean4, _ = bootstrap_mean_test(data4, m0=10, alternative='two-sided')
print("1.1.a.4 exponential mean != 10 :", p4, "| mean =", mean4)


# ================================================================
# 📘 1.1.a.5 Пример — малый объём данных (n=8)
# ================================================================
data5 = [7, 5, 6, 20, 8, 9, 3, 10]
p5, mean5, _ = bootstrap_mean_test(data5, m0=8, alternative='two-sided')
print("1.1.a.5 small sample mean != 8 :", p5, "| mean =", mean5)


# ================================================================
# 📘 1.1.a.6 Пример — сильные выбросы
# ================================================================
data6 = [10, 11, 12, 8, 9, 10, 11, 200]  # выброс
p6, mean6, _ = bootstrap_mean_test(data6, m0=12, alternative='two-sided')
print("1.1.a.6 mean != 12 (outliers) :", p6, "| mean =", mean6)


# ================================================================
# 📘 1.1.a.7 Пример — бизнес-кейс (выручка)
# ================================================================
sales = [520, 470, 480, 510, 690, 1000, 430, 500]
p7, mean7, _ = bootstrap_mean_test(sales, m0=500, alternative='two-sided')
print("1.1.a.7 revenue mean != 500 :", p7, "| mean =", mean7)


# ================================================================
# 📘 Итог:
# ------------------------------------------------
# Ты получил:
#  ✔ функцию bootstrap_mean_test
#  ✔ mean < m0        (1.1.a.1)
#  ✔ mean > m0        (1.1.a.2)
#  ✔ mean != m0       (1.1.a.3)
#  ✔ ненормальность   (1.1.a.4)
#  ✔ малые выборки    (1.1.a.5)
#  ✔ выбросы          (1.1.a.6)
#  ✔ бизнес-кейсы     (1.1.a.7)
# ================================================================
```
