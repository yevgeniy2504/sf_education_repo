# 🔵 1. ОДНА ГРУППА (сравнение с числом)

## 1.1 One-sample t-test  
**Когда:** одна выборка, нормальность, сравнение со средним значением  
**Пример:** проверить, стало ли среднее время доставки меньше 40 минут  
(одна группа, сравнение с нормативом)

### 1.1.a Bootstrap mean test  
**Когда:** нет нормальности, выбросы  
**Пример:** оценить средний расход топлива автомобиля при "грязных" данных

### 1.1.b Permutation one-sample test  
**Когда:** вообще не хотим предположений  
**Пример:** проверить, превышает ли средняя зарплата целевой показатель

### 1.1.c One-sample Wilcoxon signed-rank  
**Когда:** ненормальность + симметрия  
**Пример:** сравнить медианное время реакции оператора с нормативом

---

## 1.2 Sign-test  
**Когда:** одна выборка, ненормальные данные, проверка медианы  
**Пример:** проверить, изменилась ли медиана количества ошибок сотрудника

### 1.2.a One-sample Wilcoxon  
**Когда:** мощнее Sign-test  
**Пример:** медианное время обработки заявки относительно стандарта

### 1.2.b Median test (1-sample)  
**Когда:** сильные выбросы  
**Пример:** проверить, превышает ли медиана температуры оборудования норму

```python
from scipy import stats
```

```python
# =====================================================================
# 📘  1.1 ONE-SAMPLE T-TEST — ВСЕ СТАНДАРТНЫЕ ВАРИАНТЫ ЗАДАЧ
# =====================================================================
# ДАННЫЕ: одна группа чисел
data = [31, 38, 30, 46, 40, 36, 38, 44]

# НОРМАТИВ / ЦЕЛЕВОЕ ЧИСЛО
norm = 40


# =====================================================================
# 1) ПРОВЕРИТЬ: СРЕДНЕЕ МЕНЬШЕ НОРМЫ  (mean < norm)
# =====================================================================
# Пример задачи:
# • время доставки < 40?
# • температура оборудования ниже нормы?
# • средние продажи упали?
#
# H0: mean = norm
# H1: mean < norm
result_less = stats.ttest_1samp(data, popmean=norm, alternative='less')
print("1) mean < norm :", result_less)


# =====================================================================
# 2) ПРОВЕРИТЬ: СРЕДНЕЕ БОЛЬШЕ НОРМЫ  (mean > norm)
# =====================================================================
# Пример задачи:
# • производительность выросла > стандарта?
# • средняя зарплата выше 300 000?
#
# H0: mean = norm
# H1: mean > norm
result_greater = stats.ttest_1samp(data, popmean=norm, alternative='greater')
print("2) mean > norm :", result_greater)


# =====================================================================
# 3) ПРОВЕРИТЬ: СРЕДНЕЕ НЕ РАВНО НОРМЕ  (mean != norm)
# =====================================================================
# Пример задачи:
# • потребление энергии изменилось? (неважно, больше или меньше)
# • средняя длительность звонка изменилась?
#
# H0: mean = norm
# H1: mean != norm
result_two_sided = stats.ttest_1samp(data, popmean=norm, alternative='two-sided')
print("3) mean != norm :", result_two_sided)


# =====================================================================
# 4) ПРОВЕРИТЬ: СРЕДНЕЕ РАВНО НОРМЕ  (mean = norm)
# =====================================================================
# ВАЖНО:
# В t-test проверка "равно" = это всегда H0.
# Специального вызова НЕТ.
#
# Интерпретация:
# • если p-value большое → НЕТ оснований отвергнуть mean = norm
# • если p-value маленькое → среднее ≠ norm
#
print("4) Проверка равенства — это интерпретация результата теста выше")


# =====================================================================
# 5) ПРОВЕРИТЬ: МЕДИАНУ (если данные НЕ нормальные)
# =====================================================================
# Для медианы используется НЕ t-test:
#   • Sign-test
#   • Wilcoxon signed-rank (если симметрия)
#
# Проверить: median < norm

n_nonzero = sum(1 for x in data if x != norm)
if n_nonzero == 0:
    raise ValueError("All observations equal to norm; sign test is undefined")

n_positive = sum(1 for x in data if x > norm)
result_median = stats.binomtest(n_positive, n_nonzero, p=0.5, alternative='two-sided')
print("5) median != norm (Sign-test):", result_median)

# Примеры задач:
# • медианное время реакции ≠ стандарту
# • медианное время обработки заявок изменилось
# • медианная зарплата ≠ целевому уровню


# =====================================================================
# 6) ПРИМЕРЫ РЕАЛЬНЫХ ЗАДАЧ ДЛЯ ONE-SAMPLE T-TEST
# =====================================================================
# • Время доставки < нормы?
# • Средняя температура > допустимого порога?
# • Средняя производительность выросла > стандарта?
# • Средняя длительность звонка ≠ прошлому месяцу?
# • Средний срок ремонта = регламенту?
# • Средняя зарплата > установленного минимума?
# • Средний расход топлива < паспортного значения?
#
# Для всех этих задач используется t-test, если данные нормальные
# и альтернатива ВСЕГДА задается через:
#   alternative='less' / 'greater' / 'two-sided'
#
# =====================================================================
```

```python
# ================================================================
# 📘 1.1.a ФУНКЦИЯ: Bootstrap Mean Test
# ================================================================
# Bootstrap-тест на среднее применяется когда:
#   • данные НЕ нормальные
#   • есть выбросы
#   • малый размер выборки (n < 30)
#   • нужно тестировать ИМЕННО СРЕДНЕЕ, а не медиану
#
# Проверяем:
#   H0: mean(data) = m0
#   H1: mean(data) < m0   (alternative='less')
#   H1: mean(data) > m0   (alternative='greater')
#   H1: mean(data) != m0  (alternative='two-sided')
#
# ================================================================

import numpy as np


def bootstrap_mean_test(data, m0, alternative='two-sided', n_boot=20000):
    """
    Bootstrap проверка среднего.
    data — список или numpy массив
    m0 — число, с которым сравниваем среднее
    alternative — 'less', 'greater', 'two-sided'
    """
    data = np.array(data)
    n = len(data)

    # Bootstrap выборки
    boot_means = np.array([
        np.mean(np.random.choice(data, size=n, replace=True))
        for _ in range(n_boot)
    ])

    # Проверка гипотез
    if alternative == 'less':
        p = np.mean(boot_means < m0)

    elif alternative == 'greater':
        p = np.mean(boot_means > m0)

    elif alternative == 'two-sided':
        diff = abs(np.mean(data) - m0)
        p = np.mean(abs(boot_means - m0) >= diff)

    else:
        raise ValueError("alternative must be 'less', 'greater', or 'two-sided'")

    return p, np.mean(data), boot_means


# ================================================================
# 📘 1.1.a.1 Пример — mean < m0
# ================================================================
# Задача: среднее время доставки МЕНЬШЕ 40?
data1 = [31, 38, 30, 46, 40, 36, 38, 44]
p1, mean1, _ = bootstrap_mean_test(data1, m0=40, alternative='less')
print("1.1.a.1 mean < 40 :", p1, "| mean =", mean1)


# ================================================================
# 📘 1.1.a.2 Пример — mean > m0
# ================================================================
# Задача: производительность БОЛЬШЕ 100?
data2 = [102, 99, 110, 108, 112, 115, 98]
p2, mean2, _ = bootstrap_mean_test(data2, m0=100, alternative='greater')
print("1.1.a.2 mean > 100 :", p2, "| mean =", mean2)


# ================================================================
# 📘 1.1.a.3 Пример — mean != m0 (двусторонний)
# ================================================================
# Задача: изменилось ли среднее потребление энергии ≠ 500?
data3 = [480, 520, 505, 499, 630, 520, 480, 700]
p3, mean3, _ = bootstrap_mean_test(data3, m0=500, alternative='two-sided')
print("1.1.a.3 mean != 500 :", p3, "| mean =", mean3)


# ================================================================
# 📘 1.1.a.4 Пример — ненормальные данные (экспонента)
# ================================================================
data4 = np.random.exponential(scale=10, size=50)
p4, mean4, _ = bootstrap_mean_test(data4, m0=10, alternative='two-sided')
print("1.1.a.4 exponential mean != 10 :", p4, "| mean =", mean4)


# ================================================================
# 📘 1.1.a.5 Пример — малый объём данных (n=8)
# ================================================================
data5 = [7, 5, 6, 20, 8, 9, 3, 10]
p5, mean5, _ = bootstrap_mean_test(data5, m0=8, alternative='two-sided')
print("1.1.a.5 small sample mean != 8 :", p5, "| mean =", mean5)


# ================================================================
# 📘 1.1.a.6 Пример — сильные выбросы
# ================================================================
data6 = [10, 11, 12, 8, 9, 10, 11, 200]  # выброс
p6, mean6, _ = bootstrap_mean_test(data6, m0=12, alternative='two-sided')
print("1.1.a.6 mean != 12 (outliers) :", p6, "| mean =", mean6)


# ================================================================
# 📘 1.1.a.7 Пример — бизнес-кейс (выручка)
# ================================================================
sales = [520, 470, 480, 510, 690, 1000, 430, 500]
p7, mean7, _ = bootstrap_mean_test(sales, m0=500, alternative='two-sided')
print("1.1.a.7 revenue mean != 500 :", p7, "| mean =", mean7)


# ================================================================
# 📘 Итог:
# ------------------------------------------------
# Ты получил:
#  ✔ функцию bootstrap_mean_test
#  ✔ mean < m0        (1.1.a.1)
#  ✔ mean > m0        (1.1.a.2)
#  ✔ mean != m0       (1.1.a.3)
#  ✔ ненормальность   (1.1.a.4)
#  ✔ малые выборки    (1.1.a.5)
#  ✔ выбросы          (1.1.a.6)
#  ✔ бизнес-кейсы     (1.1.a.7)
# ================================================================
```

```python
# ================================================================
# 📘 1.1.b ФУНКЦИЯ: Permutation One-sample Test
# ================================================================
# Пермутационный тест — это альтернатива t-test,
# когда мы НЕ хотим предполагать нормальность.
#
# ИДЕЯ:
#   1) считаем разницу mean(data) - m0
#   2) многократно случайно меняем знак элементов (симметризация)
#   3) вычисляем распределение разниц
#   4) считаем p-value как долю "случайных" разниц,
#      которые оказались более экстремальными, чем наблюдаемая.
#
# Используется когда:
#   • нет нормальности
#   • есть выбросы
#   • малый размер выборки
#   • хотим тестировать СРЕДНЕЕ (не медиану)
#
# Проверяем:
#   H0: mean = m0
#   H1: mean < m0   ('less')
#   H1: mean > m0   ('greater')
#   H1: mean != m0  ('two-sided')
# ================================================================

import numpy as np

def permutation_one_sample_test(data, m0, alternative='two-sided', n_perm=20000):
    data = np.array(data)
    n = len(data)

    # наблюдаемая статистика
    observed = np.mean(data) - m0

    # генерируем случайные комбинации знаков
    signs = np.random.choice([-1, 1], size=(n_perm, n))
    permuted = (data - m0) * signs
    perm_stats = permuted.mean(axis=1)

    # выбор альтернативы
    if alternative == 'less':
        p_value = np.mean(perm_stats < observed)

    elif alternative == 'greater':
        p_value = np.mean(perm_stats > observed)

    elif alternative == 'two-sided':
        p_value = np.mean(np.abs(perm_stats) >= abs(observed))

    else:
        raise ValueError("alternative must be: 'less', 'greater', 'two-sided'")

    return p_value, np.mean(data), observed


# ================================================================
# 📘 1.1.b.1 Пример — mean < m0
# ================================================================
# Задача: проверить, уменьшилось ли среднее время доставки < 40
data1 = [31, 38, 30, 46, 40, 36, 38, 44]

p1, mean1, obs1 = permutation_one_sample_test(data1, m0=40, alternative='less')
print("1.1.b.1 mean < 40 :", p1, "| mean =", mean1, "| observed =", obs1)


# ================================================================
# 📘 1.1.b.2 Пример — mean > m0
# ================================================================
# Задача: средняя производительность > 100?
data2 = [102, 99, 110, 108, 112, 115, 98]

p2, mean2, obs2 = permutation_one_sample_test(data2, m0=100, alternative='greater')
print("1.1.b.2 mean > 100 :", p2, "| mean =", mean2, "| observed =", obs2)


# ================================================================
# 📘 1.1.b.3 Пример — mean != m0 (двусторонний)
# ================================================================
# Задача: потребление энергии ≠ 500?
data3 = [480, 520, 505, 499, 630, 520, 480, 700]

p3, mean3, obs3 = permutation_one_sample_test(data3, m0=500, alternative='two-sided')
print("1.1.b.3 mean != 500 :", p3, "| mean =", mean3, "| observed =", obs3)


# ================================================================
# 📘 1.1.b.4 Пример — НЕнормальные данные (экспонента)
# ================================================================
data4 = np.random.exponential(scale=10, size=40)

p4, mean4, obs4 = permutation_one_sample_test(data4, m0=10, alternative='two-sided')
print("1.1.b.4 exponential != 10 :", p4, "| mean =", mean4, "| observed =", obs4)


# ================================================================
# 📘 1.1.b.5 Пример — маленький размер выборки (n = 7)
# ================================================================
data5 = [7, 5, 6, 20, 8, 9, 3]  # t-test противопоказан

p5, mean5, obs5 = permutation_one_sample_test(data5, m0=8, alternative='two-sided')
print("1.1.b.5 small n mean != 8 :", p5, "| mean =", mean5, "| observed =", obs5)


# ================================================================
# 📘 1.1.b.6 Пример — сильные выбросы
# ================================================================
data6 = [10, 11, 12, 8, 9, 10, 11, 200]  # выброс

p6, mean6, obs6 = permutation_one_sample_test(data6, m0=12, alternative='two-sided')
print("1.1.b.6 mean != 12 (outliers) :", p6, "| mean =", mean6, "| observed =", obs6)


# ================================================================
# 📘 1.1.b.7 Пример — бизнес-кейс (выручка магазина)
# ================================================================
sales = [520, 470, 480, 510, 690, 1000, 430, 500]
p7, mean7, obs7 = permutation_one_sample_test(sales, m0=500, alternative='two-sided')
print("1.1.b.7 revenue mean != 500 :", p7, "| mean =", mean7, "| observed =", obs7)


# ================================================================
# 📘 Итог:
# ------------------------------------------------
# В разделе 1.1.b теперь есть:
#  ✔ Полная функция permutation_one_sample_test
#  ✔ mean < m0        (1.1.b.1)
#  ✔ mean > m0        (1.1.b.2)
#  ✔ mean != m0       (1.1.b.3)
#  ✔ ненормальность   (1.1.b.4)
#  ✔ малые выборки    (1.1.b.5)
#  ✔ выбросы          (1.1.b.6)
#  ✔ бизнес-кейсы     (1.1.b.7)
# ================================================================
```

```python
# ================================================================
# 📘 1.1.c ФУНКЦИЯ: One-sample Wilcoxon Signed-Rank Test
# ================================================================
# Используется когда:
#   • одна группа данных
#   • данные НЕ нормальные
#   • распределение примерно СИММЕТРИЧНОЕ
#   • нужно сравнить МЕДИАНУ с гипотетическим значением
#
# Сравниваем:
#   H0: median(data) = m0
#   H1: median(data) < m0    ('less')
#   H1: median(data) > m0    ('greater')
#   H1: median(data) != m0   ('two-sided')
#
# ВНИМАНИЕ:
#   Стандартная реализация scipy.stats.wilcoxon
#   работает как "paired Wilcoxon", но используется
#   как one-sample, если подавать (data - m0).
#
# Wilcoxon — это более сильная альтернатива Sign-test.
# ================================================================

import numpy as np
from scipy.stats import wilcoxon


def wilcoxon_one_sample(data, m0, alternative='two-sided'):
    """
    One-sample Wilcoxon test через вычитание m0.
    """
    data = np.array(data)
    diff = data - m0

    # scipy принимает alternative = 'less', 'greater', 'two-sided'
    result = wilcoxon(diff, alternative=alternative)
    return result.pvalue, np.median(data), np.mean(diff)


# ================================================================
# 📘 1.1.c.1 Пример — median < m0
# ================================================================
# Задача: медианное время реакции < 300 мс?
data1 = [280, 290, 295, 310, 305, 275, 260]

p1, med1, _ = wilcoxon_one_sample(data1, m0=300, alternative='less')
print("1.1.c.1 median < 300 :", p1, "| median =", med1)


# ================================================================
# 📘 1.1.c.2 Пример — median > m0
# ================================================================
# Задача: медианное время обработки > 4 секунд?
data2 = [4.1, 4.5, 4.3, 4.8, 5.2, 4.0]

p2, med2, _ = wilcoxon_one_sample(data2, m0=4, alternative='greater')
print("1.1.c.2 median > 4 :", p2, "| median =", med2)


# ================================================================
# 📘 1.1.c.3 Пример — median != m0
# ================================================================
# Задача: медианное потребление энергии ≠ 120?
data3 = [100, 115, 125, 130, 110, 140]

p3, med3, _ = wilcoxon_one_sample(data3, m0=120, alternative='two-sided')
print("1.1.c.3 median != 120 :", p3, "| median =", med3)


# ================================================================
# 📘 1.1.c.4 Пример — НЕнормальные данные (экспонента)
# ================================================================
data4 = np.random.exponential(scale=10, size=30)

p4, med4, _ = wilcoxon_one_sample(data4, m0=10, alternative='two-sided')
print("1.1.c.4 exponential median != 10 :", p4, "| median =", med4)


# ================================================================
# 📘 1.1.c.5 Пример — асимметричное распределение
# ================================================================
# Wilcoxon всё ещё работает, но мощность ниже.
data5 = [2, 3, 5, 10, 15, 20, 50]   # сильная асимметрия

p5, med5, _ = wilcoxon_one_sample(data5, m0=10, alternative='two-sided')
print("1.1.c.5 skewed median != 10 :", p5, "| median =", med5)


# ================================================================
# 📘 1.1.c.6 Пример — выбросы
# ================================================================
data6 = [12, 13, 14, 15, 16, 200]  # выброс 200

p6, med6, _ = wilcoxon_one_sample(data6, m0=15, alternative='two-sided')
print("1.1.c.6 median != 15 (outlier) :", p6, "| median =", med6)


# ================================================================
# 📘 1.1.c.7 Пример — малый объём данных (n=6)
# ================================================================
data7 = [10, 11, 9, 8, 10, 12]

p7, med7, _ = wilcoxon_one_sample(data7, m0=10, alternative='two-sided')
print("1.1.c.7 small sample median != 10 :", p7, "| median =", med7)


# ================================================================
# 📘 1.1.c.8 Пример — бизнес-кейс: время обработки заявок
# ================================================================
data8 = [34, 36, 35, 33, 38, 40, 30, 28]
m0 = 35  # норматив

p8, med8, _ = wilcoxon_one_sample(data8, m0=m0, alternative='less')
print("1.1.c.8 median < 35 (processing time) :", p8, "| median =", med8)


# ================================================================
# 📘 ИТОГ (что включено):
# ------------------------------------------------
# ✔ Функция wilcoxon_one_sample()
# ✔ median < m0         (1.1.c.1)
# ✔ median > m0         (1.1.c.2)
# ✔ median != m0        (1.1.c.3)
# ✔ ненормальность      (1.1.c.4)
# ✔ асимметрия          (1.1.c.5)
# ✔ выбросы             (1.1.c.6)
# ✔ малый объём n       (1.1.c.7)
# ✔ бизнес-кейс         (1.1.c.8)
# ================================================================
```

```python
# ================================================================
# 📘 1.2 ОДНА ГРУППА — Sign-test (реальная библиотечная версия)
# ================================================================
# Используется когда:
#   • нужно проверить МЕДИАНУ одной выборки относительно числа m0
#   • данные НЕ нормальны
#   • нет симметрии
#   • есть выбросы
#   • малый объём n
#
# Ограничение:
#   sign_test() поддерживает ТОЛЬКО two-sided.
#   Нельзя делать:
#       alternative="greater"
#       alternative="less"
#
# Это реальные лимиты функции в statsmodels.
# ================================================================

import numpy as np
from statsmodels.stats.descriptivestats import sign_test


# ================================================================
# 📘 1.2.1 median(data) = m0 ?  — двусторонний тест
# ================================================================
# Пример: медиана времени загрузки = 5 секунд?

data1 = np.array([3, 4, 6, 2, 1, 5, 4])

stat1, p1 = sign_test(data1, mu0=5)
print("1.2.1 median == 5 :", p1, "| test statistic =", stat1)


# ================================================================
# 📘 1.2.2 Бизнес-кейс — медиана ошибок = 10?
# ================================================================
errors = np.array([8, 11, 9, 12, 13, 10, 6, 15])

stat2, p2 = sign_test(errors, mu0=10)
print("1.2.2 median == 10 :", p2)


# ================================================================
# 📘 1.2.3 Малый объём данных (реальная сильная сторона теста)
# ================================================================
small = np.array([5, 7, 3, 9])

stat3, p3 = sign_test(small, mu0=6)
print("1.2.3 small n, median == 6 :", p3)


# ================================================================
# 📘 1.2.4 Устойчивость к выбросам — знак, а не значение
# ================================================================
messy = np.array([10, 12, 11, 200, 13, 9])  # выбросы не влияют

stat4, p4 = sign_test(messy, mu0=12)
print("1.2.4 outliers, median == 12 :", p4)


# ================================================================
# 📘 1.2.5 Плохие данные — огромные скачки, но знак работает
# ================================================================
bad_data = np.array([0, 1000, -200, 5000, -1000, 300, -50])

stat5, p5 = sign_test(bad_data, mu0=0)
print("1.2.5 messy data, median == 0 :", p5)


# ================================================================
# 📘 1.2.6 Бизнес-кейс — заявки < 40?
# ================================================================
# ВАЖНО: Sign-test НЕ умеет alternative="less".
# Поэтому используем только проверку равенства медианы.

processing = np.array([35, 42, 44, 38, 37, 39, 41])
                    
stat6, p6 = sign_test(processing, mu0=40)
print("1.2.6 processing median == 40 :", p6)


# ================================================================
# 📘 ИТОГ:
# ------------------------------------------------
# ✔ Реальная библиотечная функция sign_test()
# ✔ Только two-sided
# ✔ Проверка median(data) = m0
# ✔ Устойчивость к выбросам
# ✔ Малые выборки
# ✔ Плохие данные
# ✔ Бизнес-кейс
# ------------------------------------------------
# Для alternative = "less" / "greater":
#   → использовать один из тестов:
#       • Wilcoxon signed-rank (если разности симметричны)
#       • Permutation test (если без предположений)
# ================================================================
```

```python
# ================================================================
# 📘 1.2.a ФУНКЦИЯ: One-sample Wilcoxon (вариант Sign-test++ )
# ================================================================
# Используется когда:
#   • хотим тестировать МЕДИАНУ
#   • данные НЕ нормальные
#   • распределение НЕ обязательно симметричное
#   • выбросы присутствуют
#   • малый объём данных допустим
#   • НО в отличие от sign-test учитываются РАНГИ разностей
#
# Проверяем:
#   H0: median(data) = m0
#   H1: median(data) < m0    ('less')
#   H1: median(data) > m0    ('greater')
#   H1: median(data) != m0   ('two-sided')
#
# Реализация через scipy.stats.wilcoxon(data - m0)
#
# Этот тест сильнее Sign-test, но слабее классического Wilcoxon Signed-Rank,
# когда нельзя использовать симметричность.
# ================================================================

import numpy as np
from scipy.stats import wilcoxon


def one_sample_wilcoxon(data, m0, alternative='two-sided'):
    data = np.array(data)
    diff = data - m0  # превращаем в paired-разности
    result = wilcoxon(diff, alternative=alternative)
    return result.pvalue, np.median(data)


# ================================================================
# 📘 1.2.a.1 median < m0
# ================================================================
# Задача: медианное время загрузки < 5 секунд?
data1 = [3, 4, 6, 2, 4, 5, 3]

p1, med1 = one_sample_wilcoxon(data1, m0=5, alternative='less')
print("1.2.a.1 median < 5 :", p1, "| median =", med1)


# ================================================================
# 📘 1.2.a.2 median > m0
# ================================================================
# Задача: медианная производительность > 100?
data2 = [110, 120, 95, 105, 130, 125]

p2, med2 = one_sample_wilcoxon(data2, m0=100, alternative='greater')
print("1.2.a.2 median > 100 :", p2, "| median =", med2)


# ================================================================
# 📘 1.2.a.3 median != m0
# ================================================================
# Задача: медианный расход энергии ≠ 500?
data3 = [480, 520, 510, 530, 495, 505]

p3, med3 = one_sample_wilcoxon(data3, m0=500, alternative='two-sided')
print("1.2.a.3 median != 500 :", p3, "| median =", med3)


# ================================================================
# 📘 1.2.a.4 НЕнормальные данные (экспонента)
# ================================================================
data4 = np.random.exponential(scale=10, size=25)

p4, med4 = one_sample_wilcoxon(data4, m0=10, alternative='two-sided')
print("1.2.a.4 exponential median != 10:", p4, "| median =", med4)


# ================================================================
# 📘 1.2.a.5 Выбросы
# ================================================================
# Wilcoxon чувствителен к РАНГАМ, выброс не ломает тест как t-test.
data5 = [10, 12, 11, 13, 400]  # выброс 400

p5, med5 = one_sample_wilcoxon(data5, m0=12, alternative='two-sided')
print("1.2.a.5 outlier median != 12 :", p5, "| median =", med5)


# ================================================================
# 📘 1.2.a.6 Малый объём данных (n = 6)
# ================================================================
data6 = [7, 8, 9, 10, 11, 12]

p6, med6 = one_sample_wilcoxon(data6, m0=10, alternative='two-sided')
print("1.2.a.6 small sample median != 10 :", p6, "| median =", med6)


# ================================================================
# 📘 1.2.a.7 Очень плохие данные (много нулевых разностей)
# ================================================================
data7 = [10, 10, 10, 12, 8, 10]

p7, med7 = one_sample_wilcoxon(data7, m0=10, alternative='two-sided')
print("1.2.a.7 messy data median != 10 :", p7, "| median =", med7)


# ================================================================
# 📘 1.2.a.8 Бизнес-кейс
# ================================================================
# Проверить: уменьшилась ли медиана времени обработки < 40 минут?
data8 = [42, 38, 37, 45, 39, 40, 35]

p8, med8 = one_sample_wilcoxon(data8, m0=40, alternative='less')
print("1.2.a.8 processing median < 40 :", p8, "| median =", med8)


# ================================================================
# 📘 ИТОГ:
# ------------------------------------------------
# ✔ Функция one_sample_wilcoxon()
# ✔ median < m0               (1.2.a.1)
# ✔ median > m0               (1.2.a.2)
# ✔ median != m0              (1.2.a.3)
# ✔ НЕнормальные данные       (1.2.a.4)
# ✔ Выбросы                   (1.2.a.5)
# ✔ Малое n                   (1.2.a.6)
# ✔ "Плохие" данные           (1.2.a.7)
# ✔ Бизнес-кейс               (1.2.a.8)
# ================================================================
```

```python
# ================================================================
# 📘 1.2.b ФУНКЦИЯ: One-sample Median Test
# ================================================================
# ЛОГИКА:
#   Проверяем, отличается ли МЕДИАНА выборки от m0,
#   сравнивая количество элементов выше и ниже m0.
#
# Это НЕ sign-test, но математически
# это тот же биномиальный тест, НО интерпретируется как median test.
#
# Использовать, когда:
#   • данные не нормальные
#   • распределение НЕ симметричное
#   • критично проверять именно медиану
#   • есть выбросы
#   • маленькая выборка
#
# Проверяем:
#   H0: median = m0
#   H1: median < m0
#   H1: median > m0
#   H1: median != m0
# ================================================================

import numpy as np
from math import comb

def median_test_one_sample(data, m0, alternative="two-sided"):
    data = np.array(data)
    pos = np.sum(data > m0)
    neg = np.sum(data < m0)
    n = pos + neg

    if n == 0:
        return 1.0, np.median(data)  # все элементы равны m0

    # биномиальный тест B(n, 0.5)
    def p_val(k, n, alt):
        p = 0.0
        if alt == "greater":
            for i in range(k, n + 1):
                p += comb(n, i) * 0.5**n
        elif alt == "less":
            for i in range(0, k + 1):
                p += comb(n, i) * 0.5**n
        elif alt == "two-sided":
            tail = min(k, n - k)
            for i in range(0, tail + 1):
                p += comb(n, i) * 0.5**n
        return p

    p = p_val(pos, n, alternative)
    return p, np.median(data)


# ================================================================
# 📘 1.2.b.1 median < m0
# ================================================================
# Пример: медианная задержка < 100 мс?
data1 = [80, 120, 90, 70, 110, 95]

p1, med1 = median_test_one_sample(data1, m0=100, alternative="less")
print("1.2.b.1 median < 100 :", p1, "| median =", med1)


# ================================================================
# 📘 1.2.b.2 median > m0
# ================================================================
# Пример: медианный доход > 2000?
data2 = [1500, 2100, 2500, 1900, 2300, 2200]

p2, med2 = median_test_one_sample(data2, m0=2000, alternative="greater")
print("1.2.b.2 median > 2000 :", p2, "| median =", med2)


# ================================================================
# 📘 1.2.b.3 median != m0 (двусторонний)
# ================================================================
# Пример: медианное давление ≠ 120?
data3 = [118, 122, 130, 110, 125, 117]

p3, med3 = median_test_one_sample(data3, m0=120, alternative="two-sided")
print("1.2.b.3 median != 120 :", p3, "| median =", med3)


# ================================================================
# 📘 1.2.b.4 НЕнормальные данные (экспонента)
# ================================================================
data4 = np.random.exponential(scale=5, size=40)

p4, med4 = median_test_one_sample(data4, m0=5, alternative="two-sided")
print("1.2.b.4 exponential median != 5 :", p4, "| median =", med4)


# ================================================================
# 📘 1.2.b.5 Выбросы
# ================================================================
data5 = [10, 10, 11, 12, 500, 9, 8]

p5, med5 = median_test_one_sample(data5, m0=12, alternative="two-sided")
print("1.2.b.5 outliers median != 12 :", p5, "| median =", med5)


# ================================================================
# 📘 1.2.b.6 Малый объём (n = 5)
# ================================================================
data6 = [5, 7, 9, 11, 13]

p6, med6 = median_test_one_sample(data6, m0=10, alternative="two-sided")
print("1.2.b.6 small sample median != 10 :", p6, "| median =", med6)


# ================================================================
# 📘 1.2.b.7 Бизнес-кейс
# ================================================================
# Пример: медианная длительность звонка снизилась < 4 минуты?
calls = [5, 4, 3, 6, 4, 3, 2]

p7, med7 = median_test_one_sample(calls, m0=4, alternative="less")
print("1.2.b.7 call duration median < 4 :", p7, "| median =", med7)


# ================================================================
# 📘 ИТОГ:
# ------------------------------------------------
# ✔ Реализован One-sample Median Test
# ✔ median < m0              (1.2.b.1)
# ✔ median > m0              (1.2.b.2)
# ✔ median != m0             (1.2.b.3)
# ✔ Ненормальность           (1.2.b.4)
# ✔ Выбросы                  (1.2.b.5)
# ✔ Малое n                  (1.2.b.6)
# ✔ Бизнес-кейс              (1.2.b.7)
# ================================================================
```

```python
# ================================================================
# 📘 2.1 ФУНКЦИЯ: Independent t-test (две независимые группы)
# ================================================================
# Используется когда:
#   • сравниваем СРЕДНИЕ двух независимых групп
#   • данные примерно нормальные
#   • дисперсии могут быть:
#       - равны -> classical t-test
#       - не равны -> Welch t-test
#
# SciPy:
#   stats.ttest_ind(x, y, equal_var=True/False, alternative='two-sided')
#
# Проверяем:
#   H0: mean₁ = mean₂
#   H1: mean₁ < mean₂
#   H1: mean₁ > mean₂
#   H1: mean₁ ≠ mean₂
#
# Частые применения:
#   • сравнение среднего времени выполнения задачи в двух отделах
#   • сравнение средних продаж в двух магазинах
#   • сравнение уровня успеваемости в двух школах
# ================================================================

import numpy as np
from scipy import stats


def independent_ttest(g1, g2, alternative="two-sided", equal_var=True):
    """
    g1, g2       — две независимые группы
    equal_var    — True: классический t-test
                   False: Welch t-test
    alternative  — 'less', 'greater', 'two-sided'
    """
    result = stats.ttest_ind(g1, g2, equal_var=equal_var, alternative=alternative)
    return result.pvalue, np.mean(g1), np.mean(g2)


# ================================================================
# 📘 2.1.1 mean₁ < mean₂
# ================================================================
# Задача: время обработки заявок в отделе A < отдела B?
A = [40, 42, 38, 37, 41, 39]
B = [45, 48, 47, 46, 50, 44]

p1, mA1, mB1 = independent_ttest(A, B, alternative="less")
print("2.1.1 mean_A < mean_B :", p1, "| means:", mA1, mB1)


# ================================================================
# 📘 2.1.2 mean₁ > mean₂
# ================================================================
# Задача: средний доход клиентов premium > basic?
premium  = [1200, 1500, 1700, 1800, 1600]
basic    = [900, 1100, 1000, 950, 1050]

p2, mp, mb = independent_ttest(premium, basic, alternative="greater")
print("2.1.2 mean_premium > mean_basic :", p2, "| means:", mp, mb)


# ================================================================
# 📘 2.1.3 mean₁ ≠ mean₂ (двусторонний)
# ================================================================
# Пример: средняя оценка сервиса отличается между городами?
city1 = [4.1, 4.3, 4.0, 4.2]
city2 = [3.8, 3.9, 4.0, 3.7]

p3, m1_3, m2_3 = independent_ttest(city1, city2, alternative="two-sided")
print("2.1.3 mean_city1 != mean_city2 :", p3, "| means:", m1_3, m2_3)


# ================================================================
# 📘 2.1.4 Разные дисперсии (Welch)
# ================================================================
# Пример: отделы работают по-разному → unequal variance
G1 = [10, 12, 11, 9, 8]
G2 = [20, 30, 25, 22, 28]   # большая дисперсия

p4, m41, m42 = independent_ttest(G1, G2, equal_var=False)
print("2.1.4 Welch test mean1 != mean2 :", p4, "| means:", m41, m42)


# ================================================================
# 📘 2.1.5 Равные дисперсии (Classical t-test)
# ================================================================
G3 = [100, 102, 98, 101, 99]
G4 = [103, 104, 102, 105, 101]

p5, m51, m52 = independent_ttest(G3, G4, equal_var=True)
print("2.1.5 Classical t-test:", p5, "| means:", m51, m52)


# ================================================================
# 📘 2.1.6 Малый объём данных
# ================================================================
g_small1 = [7, 8, 9]
g_small2 = [12, 13, 11]

p6, ms1, ms2 = independent_ttest(g_small1, g_small2, alternative="two-sided")
print("2.1.6 small sample mean != :", p6, "| means:", ms1, ms2)


# ================================================================
# 📘 2.1.7 Выбросы (нежелательны для t-test)
# ================================================================
# t-test чувствителен к выбросам → лучше Welch или Mann–Whitney
g_out1 = [10, 11, 12, 10, 500]   # выброс 500
g_out2 = [9, 8, 11, 10, 12]

p7, mo1, mo2 = independent_ttest(g_out1, g_out2, equal_var=False)
print("2.1.7 outliers Welch mean != :", p7, "| means:", mo1, mo2)


# ================================================================
# 📘 2.1.8 Ненормальность (t-test работает плохо)
# ================================================================
g_exp1 = np.random.exponential(scale=2, size=30)
g_exp2 = np.random.exponential(scale=4, size=30)

p8, me1, me2 = independent_ttest(g_exp1, g_exp2, equal_var=False)
print("2.1.8 exponential Welch mean != :", p8, "| means:", me1, me2)


# ================================================================
# 📘 2.1.9 Бизнес-кейс: сравнение магазинов
# ================================================================
# Средняя выручка магазина A против магазина B
store_A = [54000, 56000, 58000, 60000, 55000]
store_B = [50000, 52000, 51000, 53000, 52500]

p9, mA9, mB9 = independent_ttest(store_A, store_B, alternative="greater")
print("2.1.9 revenue_A > revenue_B :", p9, "| means:", mA9, mB9)


# ================================================================
# 📘 2.1.10 Бизнес-кейс: сравнение производительности
# ================================================================
team1 = [120, 125, 130, 128, 127]
team2 = [115, 118, 116, 117, 119]

p10, t1, t2 = independent_ttest(team1, team2, alternative="greater")
print("2.1.10 team1 > team2 :", p10, "| means:", t1, t2)


# ================================================================
# 📘 ИТОГ:
# ------------------------------------------------
# ✔ mean₁ < mean₂              (2.1.1)
# ✔ mean₁ > mean₂              (2.1.2)
# ✔ mean₁ ≠ mean₂              (2.1.3)
# ✔ Welch test (не равные var) (2.1.4)
# ✔ Classical t-test           (2.1.5)
# ✔ Малое n                    (2.1.6)
# ✔ Выбросы (лучше Welch)      (2.1.7)
# ✔ Ненормальность             (2.1.8)
# ✔ Бизнес-кейсы               (2.1.9–2.1.10)
# ================================================================
```

```python
# ================================================================
# 📘 2.1.a ФУНКЦИЯ: Welch Independent t-test
# ================================================================
# Welch t-test:
#  • сравнение СРЕДНИХ двух независимых групп
#  • НЕ требует равенства дисперсий
#  • работает даже когда дисперсии различаются в 3–10 раз
#  • менее чувствителен к выбросам, чем классический t-test
#
# Проверяем гипотезы:
#   H0: mean₁ = mean₂
#   H1: mean₁ < mean₂
#   H1: mean₁ > mean₂
#   H1: mean₁ ≠ mean₂
#
# SciPy:
#   stats.ttest_ind(x, y, equal_var=False, alternative="...")
#
# Применение:
#  • сравнение продаж магазинов с разной волатильностью
#  • сравнение времени выполнения задач между командами
#  • сравнение доходов, когда у одной группы большая дисперсия
# ================================================================

import numpy as np
from scipy import stats

def welch_ttest(g1, g2, alternative="two-sided"):
    """
    g1, g2 — две независимые группы
    alternative — 'less', 'greater', 'two-sided'
    """
    result = stats.ttest_ind(g1, g2, equal_var=False, alternative=alternative)
    return result.pvalue, np.mean(g1), np.mean(g2), np.var(g1, ddof=1), np.var(g2, ddof=1)


# ================================================================
# 📘 2.1.a.1 mean₁ < mean₂
# ================================================================
# Пример: отдел А работает быстрее отдела B?
A = [40, 38, 42, 39, 37, 41]
B = [47, 50, 49, 48, 51, 52]

p1, mA1, mB1, vA1, vB1 = welch_ttest(A, B, alternative="less")
print("2.1.a.1 mean_A < mean_B :", p1, "| means:", mA1, mB1, "| variances:", vA1, vB1)


# ================================================================
# 📘 2.1.a.2 mean₁ > mean₂
# ================================================================
# Пример: производительность команды X > команды Y?
X = [120, 125, 130, 140, 110]
Y = [90, 100, 95, 85, 105]

p2, mX2, mY2, vX2, vY2 = welch_ttest(X, Y, alternative="greater")
print("2.1.a.2 mean_X > mean_Y :", p2, "| means:", mX2, mY2, "| variances:", vX2, vY2)


# ================================================================
# 📘 2.1.a.3 mean₁ ≠ mean₂
# ================================================================
# Пример: сравнение двух городов по времени доставки
city1 = [32, 34, 36, 40]
city2 = [45, 47, 42, 50]

p3, c1m, c2m, c1v, c2v = welch_ttest(city1, city2, alternative="two-sided")
print("2.1.a.3 mean_city1 != mean_city2 :", p3, "| means:", c1m, c2m)


# ================================================================
# 📘 2.1.a.4 Очень разные дисперсии
# ================================================================
# Отличный кейс для Welch — когда одна группа "стабильная", вторая "рваная"
stable = [100, 102, 101, 99, 98]
volatile = [50, 200, 150, 180, 20]

p4, ms, mv, vs, vv = welch_ttest(stable, volatile)
print("2.1.a.4 unequal variances:", p4, "| vars:", vs, vv)


# ================================================================
# 📘 2.1.a.5 Выбросы
# ================================================================
g1 = [10, 11, 12, 10, 500]  # выброс
g2 = [8, 9, 11, 10, 12]

p5, m51, m52, v51, v52 = welch_ttest(g1, g2)
print("2.1.a.5 outlier case:", p5, "| means:", m51, m52)


# ================================================================
# 📘 2.1.a.6 Малый объём данных
# ================================================================
g_small1 = [7, 8, 9]
g_small2 = [11, 12, 10]

p6, s1, s2, sv1, sv2 = welch_ttest(g_small1, g_small2)
print("2.1.a.6 small n:", p6, "| means:", s1, s2)


# ================================================================
# 📘 2.1.a.7 Ненормальность (экспонента)
# ================================================================
exp1 = np.random.exponential(scale=5, size=20)
exp2 = np.random.exponential(scale=10, size=20)

p7, e1, e2, ev1, ev2 = welch_ttest(exp1, exp2)
print("2.1.a.7 exponential:", p7, "| means:", e1, e2)


# ================================================================
# 📘 2.1.a.8 Бизнес-кейс — магазины
# ================================================================
# Магазин A работает стабильнее, B — нестабильно
store_A = [52000, 53000, 51000, 52500, 51500]
store_B = [48000, 62000, 45000, 59000, 47000]

p8, A8, B8, vA8, vB8 = welch_ttest(store_A, store_B, alternative="greater")
print("2.1.a.8 revenue_A > revenue_B:", p8, "| means:", A8, B8)


# ================================================================
# 📘 2.1.a.9 Бизнес-кейс — производительность команд
# ================================================================
team_fast = [120, 130, 110, 125, 140, 150]
team_slow = [80, 90, 85, 92, 88, 95]

p9, f9, s9, vf9, vs9 = welch_ttest(team_fast, team_slow, alternative="greater")
print("2.1.a.9 team_fast > team_slow:", p9, "| means:", f9, s9)


# ================================================================
# 📘 ИТОГ:
# ------------------------------------------------
# ✔ mean₁ < mean₂                     (2.1.a.1)
# ✔ mean₁ > mean₂                     (2.1.a.2)
# ✔ mean₁ ≠ mean₂                     (2.1.a.3)
# ✔ Очень разные дисперсии            (2.1.a.4)
# ✔ Выбросы                           (2.1.a.5)
# ✔ Малый объём                       (2.1.a.6)
# ✔ Ненормальность                    (2.1.a.7)
# ✔ Бизнес-кейсы                      (2.1.a.8–9)
# ================================================================
```

```python
# ================================================================
# 📘 2.1.b ФУНКЦИЯ: Permutation 2-sample Mean Test
# ================================================================
# Используется когда:
#   • хотим сравнить СРЕДНИЕ двух независимых групп
#   • нет нормальности
#   • разные дисперсии
#   • много выбросов
#   • малые выборки (n < 20)
#   • t-test/Welch не подходят
#
# Идея:
#   1. Вычисляем разницу mean(g1) - mean(g2)
#   2. Перемешиваем группы (shuffle)
#   3. Считаем разницу средних в каждой перестановке
#   4. p-value = доля перестановок, дающих эффект ≥ наблюдаемого
#
# Проверяем гипотезы:
#   H0: mean₁ = mean₂
#   H1: mean₁ < mean₂
#   H1: mean₁ > mean₂
#   H1: mean₁ ≠ mean₂
#
# ================================================================

import numpy as np

def permutation_test_2sample(g1, g2, n_perm=20000, alternative="two-sided"):
    g1 = np.array(g1)
    g2 = np.array(g2)

    observed = g1.mean() - g2.mean()

    combined = np.concatenate([g1, g2])
    n1 = len(g1)

    perm_diffs = np.zeros(n_perm)

    for i in range(n_perm):
        perm = np.random.permutation(combined)
        perm_g1 = perm[:n1]
        perm_g2 = perm[n1:]
        perm_diffs[i] = perm_g1.mean() - perm_g2.mean()

    if alternative == "greater":
        p = np.mean(perm_diffs >= observed)

    elif alternative == "less":
        p = np.mean(perm_diffs <= observed)

    elif alternative == "two-sided":
        p = np.mean(np.abs(perm_diffs) >= abs(observed))

    return p, g1.mean(), g2.mean(), observed


# ================================================================
# 📘 2.1.b.1 mean₁ < mean₂
# ================================================================
A = [40, 42, 38, 41, 37]
B = [50, 47, 48, 52, 49]

p1, mA1, mB1, obs1 = permutation_test_2sample(A, B, alternative="less")
print("2.1.b.1 mean_A < mean_B :", p1, "| means:", mA1, mB1, "| observed:", obs1)


# ================================================================
# 📘 2.1.b.2 mean₁ > mean₂
# ================================================================
premium = [1200, 1300, 1500, 1700]
basic   = [900, 1000, 1100, 950]

p2, mP2, mB2, obs2 = permutation_test_2sample(premium, basic, alternative="greater")
print("2.1.b.2 mean_premium > mean_basic :", p2, "| means:", mP2, mB2)


# ================================================================
# 📘 2.1.b.3 mean₁ ≠ mean₂
# ================================================================
city1 = [33, 35, 37, 39, 40]
city2 = [45, 47, 50, 42, 48]

p3, m13, m23, obs3 = permutation_test_2sample(city1, city2, alternative="two-sided")
print("2.1.b.3 mean_city1 != mean_city2 :", p3, "| means:", m13, m23)


# ================================================================
# 📘 2.1.b.4 Ненормальные данные (экспонента)
# ================================================================
g_exp1 = np.random.exponential(scale=4, size=30)
g_exp2 = np.random.exponential(scale=8, size=30)

p4, e1, e2, obs4 = permutation_test_2sample(g_exp1, g_exp2)
print("2.1.b.4 exponential != :", p4, "| means:", e1, e2)


# ================================================================
# 📘 2.1.b.5 Огромные выбросы
# ================================================================
g_out1 = [10, 12, 11, 10, 500]  # выброс
g_out2 = [8, 9, 11, 10, 12]

p5, o1, o2, obs5 = permutation_test_2sample(g_out1, g_out2)
print("2.1.b.5 outliers != :", p5, "| means:", o1, o2)


# ================================================================
# 📘 2.1.b.6 Малые выборки (n = 4)
# ================================================================
g_small1 = [7, 8, 9, 10]
g_small2 = [12, 11, 13, 14]

p6, s1, s2, obs6 = permutation_test_2sample(g_small1, g_small2)
print("2.1.b.6 small n:", p6, "| means:", s1, s2)


# ================================================================
# 📘 2.1.b.7 Большие различия дисперсий
# ================================================================
stable    = [100, 102, 101, 99, 100]
unstable  = [50, 200, 150, 80, 300]

p7, st1, st2, obs7 = permutation_test_2sample(stable, unstable)
print("2.1.b.7 unequal variances:", p7, "| means:", st1, st2)


# ================================================================
# 📘 2.1.b.8 Бизнес-кейс — магазины
# ================================================================
store_A = [55000, 56000, 57000, 54000, 58000]
store_B = [50000, 52000, 51000, 49500, 50500]

p8, A8, B8, obs8 = permutation_test_2sample(store_A, store_B, alternative="greater")
print("2.1.b.8 revenue_A > revenue_B :", p8, "| means:", A8, B8)


# ================================================================
# 📘 2.1.b.9 Бизнес-кейс — производительность команд
# ================================================================
team_fast = [120, 130, 140, 150, 160]
team_slow = [90, 95, 100, 105, 110]

p9, f9, s9, obs9 = permutation_test_2sample(team_fast, team_slow, alternative="greater")
print("2.1.b.9 team_fast > team_slow :", p9, "| means:", f9, s9)


# ================================================================
# 📘 ИТОГ:
# ------------------------------------------------
# ✔ mean₁ < mean₂                 (2.1.b.1)
# ✔ mean₁ > mean₂                 (2.1.b.2)
# ✔ mean₁ ≠ mean₂                 (2.1.b.3)
# ✔ ненормальность                (2.1.b.4)
# ✔ выбросы                       (2.1.b.5)
# ✔ малое n                       (2.1.b.6)
# ✔ разные дисперсии              (2.1.b.7)
# ✔ бизнес-кейсы                  (2.1.b.8–9)
# ------------------------------------------------
# Permutation test — лучший универсальный тест на разницу средних,
# когда t-test/Welch неприменим.
# ================================================================
```

```python
# ================================================================
# 📘 2.1.c Kolmogorov–Smirnov 2-sample test
# ================================================================
# Используется когда:
#   • хотим проверить, отличаются ли ДВА РАСПРЕДЕЛЕНИЯ полностью
#   • НЕ только средние, как в t-test
#   • работает в любых условиях (не требует нормальности)
#   • может ловить:
#       - различие в форме распределений
#       - различие в дисперсиях
#       - различие в медианах
#       - различие в хвостах
#
# Проверяет:
#   H0: F₁(x) = F₂(x)   (распределения одинаковы)
#   H1: F₁(x) ≠ F₂(x)   (любое отличие)
#
# Важное:
#   • KS тест НЕ предназначен специально для сравнения средних.
#   • Он тестирует ПОЛНОЕ распределение.
# ================================================================

import numpy as np
from scipy.stats import ks_2samp


def ks_test_2sample(g1, g2):
    """
    g1, g2 — две независимые выборки
    Возвращает: p-value, максимальное отклонение D
    """
    result = ks_2samp(g1, g2, alternative="two-sided", mode="auto")
    return result.pvalue, result.statistic, np.mean(g1), np.mean(g2)


# ================================================================
# 📘 2.1.c.1 Распределения отличаются
# ================================================================
g1 = np.random.normal(0, 1, 200)
g2 = np.random.normal(2, 1, 200)  # сдвиг на 2

p1, D1, m1_1, m1_2 = ks_test_2sample(g1, g2)
print("2.1.c.1 different distributions:", p1, "| D:", D1, "| means:", m1_1, m1_2)


# ================================================================
# 📘 2.1.c.2 Средние равны, но форма распределения другая
# ================================================================
g3 = np.random.normal(0, 1, 300)
g4 = np.random.exponential(scale=1, size=300) - 1  # другое распределение, но среднее ~0

p2, D2, m2_1, m2_2 = ks_test_2sample(g3, g4)
print("2.1.c.2 same mean, different shape:", p2, "| D:", D2, "| means:", m2_1, m2_2)


# ================================================================
# 📘 2.1.c.3 Большая разница дисперсий (одно широкое, другое узкое)
# ================================================================
narrow = np.random.normal(0, 0.5, 300)
wide   = np.random.normal(0, 3.0, 300)

p3, D3, m3_1, m3_2 = ks_test_2sample(narrow, wide)
print("2.1.c.3 variance difference:", p3, "| D:", D3)


# ================================================================
# 📘 2.1.c.4 Выбросы в одной группе
# ================================================================
clean = [10, 11, 9, 10, 12, 11, 10, 9]
messy = [10, 11, 500, 9, 8, 12, 300]

p4, D4, m4_1, m4_2 = ks_test_2sample(clean, messy)
print("2.1.c.4 outliers:", p4, "| D:", D4, "| means:", m4_1, m4_2)


# ================================================================
# 📘 2.1.c.5 Маленькое n = 6
# ================================================================
small1 = [1, 2, 3, 4, 5, 6]
small2 = [10, 9, 8, 7, 6, 5]

p5, D5, m5_1, m5_2 = ks_test_2sample(small1, small2)
print("2.1.c.5 small sample:", p5, "| D:", D5)


# ================================================================
# 📘 2.1.c.6 Бизнес-кейс — сравнение поведения клиентов
# ================================================================
# Например, распределение суммы покупок до и после рекламной кампании
before = np.random.exponential(scale=70, size=500)
after  = np.random.exponential(scale=100, size=500)

p6, D6, m6_1, m6_2 = ks_test_2sample(before, after)
print("2.1.c.6 customers behavior:", p6, "| D:", D6, "| means:", m6_1, m6_2)


# ================================================================
# 📘 2.1.c.7 Бизнес-кейс — сравнение распределения времени обработки
# ================================================================
team_A = np.random.gamma(shape=2, scale=10, size=400)
team_B = np.random.gamma(shape=5, scale=6, size=400)

p7, D7, m7_1, m7_2 = ks_test_2sample(team_A, team_B)
print("2.1.c.7 processing distribution:", p7, "| D:", D7)


# ================================================================
# 📘 ИТОГ:
# ------------------------------------------------
# ✔ Отличие распределений                 (2.1.c.1)
# ✔ Средние одинаковые, форма разная      (2.1.c.2)
# ✔ Разные дисперсии                      (2.1.c.3)
# ✔ Выбросы                               (2.1.c.4)
# ✔ Малое n                               (2.1.c.5)
# ✔ Бизнес-кейс: покупатели               (2.1.c.6)
# ✔ Бизнес-кейс: скорость обработки       (2.1.c.7)
# ------------------------------------------------
# Kolmogorov–Smirnov — мощный тест на сравнение ПОЛНЫХ распределений.
# ================================================================
```

```python
# ================================================================
# 📘 2.1.d ФУНКЦИЯ: Brunner–Munzel 2-sample test
# ================================================================
# Используется когда:
#   • хотим сравнить две независимые группы
#   • данные НЕ нормальные
#   • дисперсии НЕ равны
#   • формы распределений НЕ одинаковые
#   • Mann–Whitney НЕ подходит (assumption fails)
#
# Проверяет:
#   H0: P(X < Y) = 0.5       (вероятность превосходства равна)
#   H1: P(X < Y) ≠ 0.5       (двусторонний)
#   или 'less' / 'greater'
#
# SciPy есть, но в statsmodels реализация лучше:
#   from statsmodels.stats.nonparametric import brunner_munzel
# ================================================================

import numpy as np
from scipy.stats import t

def brunner_munzel_test(x, y, alternative="two-sided"):
    """
    Realization of Brunner–Munzel test without statsmodels.
    Works on Python 3.12–3.13.
    """
    x = np.asarray(x)
    y = np.asarray(y)

    nx = len(x)
    ny = len(y)

    # ранги объединённых данных
    ranks = np.argsort(np.argsort(np.concatenate([x, y]))) + 1

    rx = ranks[:nx]
    ry = ranks[nx:]

    # средние ранги
    mean_rx = np.mean(rx)
    mean_ry = np.mean(ry)

    # параметр w
    w = (mean_rx - mean_ry) / (nx + ny)

    sx = np.sum((rx - mean_rx)**2) / (nx - 1)
    sy = np.sum((ry - mean_ry)**2) / (ny - 1)

    se = np.sqrt(sx / nx + sy / ny)

    t_stat = w / se

    # степени свободы
    df = (sx / nx + sy / ny)**2 / (
        (sx**2) / (nx**2 * (nx - 1)) +
        (sy**2) / (ny**2 * (ny - 1))
    )

    # p-value
    if alternative == "two-sided":
        p = 2 * (1 - t.cdf(abs(t_stat), df))
    elif alternative == "greater":
        p = 1 - t.cdf(t_stat, df)
    else:  # "less"
        p = t.cdf(t_stat, df)

    return t_stat, df, p


# короткое алиас-имя, как в твоём коде
bm_test = brunner_munzel_test


# ================================================================
# 📘 2.1.d.1 median₁ < median₂
# ================================================================
A = [40, 42, 38, 41, 37]
B = [47, 49, 45, 50, 52]

t1, df1, p1 = bm_test(A, B, alternative="less")
print("2.1.d.1 A < B :", p1)


# ================================================================
# 📘 2.1.d.2 median₁ > median₂
# ================================================================
premium = [1200, 1500, 1700, 1600, 1800]
basic   = [900, 1000, 1100, 950, 1050]

t2, df2, p2 = bm_test(premium, basic, alternative="greater")
print("2.1.d.2 premium > basic :", p2)


# ================================================================
# 📘 2.1.d.3 median₁ ≠ median₂
# ================================================================
city1 = [33, 35, 37, 40]
city2 = [45, 47, 42, 50]

t3, df3, p3 = bm_test(city1, city2, alternative="two-sided")
print("2.1.d.3 city1 != city2 :", p3)


# ================================================================
# 📘 2.1.d.4 Разные формы распределения
# ================================================================
expo1 = np.random.exponential(scale=3, size=200)
expo2 = np.random.exponential(scale=7, size=200)

t4, df4, p4 = bm_test(expo1, expo2)
print("2.1.d.4 different distributions :", p4)


# ================================================================
# 📘 2.1.d.5 Выбросы
# ================================================================
clean = [10, 11, 9, 10, 12, 11]
messy = [10, 12, 300, 9, 8, 500]

t5, df5, p5 = bm_test(clean, messy)
print("2.1.d.5 outliers :", p5)


# ================================================================
# 📘 2.1.d.6 Малый объём данных
# ================================================================
g_small1 = [1, 2, 3, 4]
g_small2 = [10, 9, 8, 7]

t6, df6, p6 = bm_test(g_small1, g_small2)
print("2.1.d.6 small n :", p6)


# ================================================================
# 📘 2.1.d.7 Большие различия дисперсий
# ================================================================
low_var  = np.random.normal(10, 1, 200)
high_var = np.random.normal(10, 10, 200)

t7, df7, p7 = bm_test(low_var, high_var)
print("2.1.d.7 unequal variances :", p7)


# ================================================================
# 📘 2.1.d.8 Бизнес-кейс — поведение клиентов
# ================================================================
before = np.random.gamma(shape=2, scale=20, size=400)
after  = np.random.gamma(shape=5, scale=10, size=400)

t8, df8, p8 = bm_test(before, after)
print("2.1.d.8 customers changed :", p8)


# ================================================================
# 📘 2.1.d.9 Бизнес-кейс — скорость обработки задач
# ================================================================
team_A = np.random.exponential(scale=10, size=300)
team_B = np.random.exponential(scale=5, size=300)

t9, df9, p9 = bm_test(team_A, team_B, alternative="greater")
print("2.1.d.9 team_A slower than B :", p9)



# ================================================================
# 📘 ИТОГ:
# ------------------------------------------------
# ✔ median₁ < median₂                     (2.1.d.1)
# ✔ median₁ > median₂                     (2.1.d.2)
# ✔ median₁ ≠ median₂                     (2.1.d.3)
# ✔ разные формы распределений            (2.1.d.4)
# ✔ выбросы                               (2.1.d.5)
# ✔ малые выборки                         (2.1.d.6)
# ✔ разные дисперсии                      (2.1.d.7)
# ✔ бизнес-кейсы                          (2.1.d.8–9)
# ------------------------------------------------
# Brunner–Munzel лучше Mann–Whitney когда:
#   1) дисперсии различаются
#   2) распределения имеют разную форму
#   3) наблюдения не "пересекаются"
#   4) выборки сильно отличаются по масштабу
# ================================================================
```

```python
# ================================================================
# 📘 2.2 ФУНКЦИЯ: Mann–Whitney U-test (независимые выборки)
# ================================================================
# Используется когда:
#   • сравниваем две независимые группы
#   • данные НЕ нормальные
#   • выбросы есть
#   • распределения могут быть разными, но желательно схожей формы
#
# Интерпретация:
#   • тестирует НЕ средние, а вероятность превосходства:
#       P(X > Y)
#   • если распределения одинаковой формы → эквивалентно тесту медиан
#
# Проверяет:
#   H0: распределения одинаковы
#   H1: g1 < g2     ('less')
#   H1: g1 > g2     ('greater')
#   H1: g1 != g2    ('two-sided')
#
# SciPy:
#   stats.mannwhitneyu(g1, g2, alternative="...")
# ================================================================

import numpy as np
from scipy.stats import mannwhitneyu


def mw_test(g1, g2, alternative="two-sided"):
    result = mannwhitneyu(g1, g2, alternative=alternative)
    return result.pvalue, np.median(g1), np.median(g2)


# ================================================================
# 📘 2.2.1 median₁ < median₂
# ================================================================
# Пример: время ответа отдела A < отдела B?
A = [40, 41, 39, 37, 38]
B = [50, 52, 48, 46, 49]

p1, medA1, medB1 = mw_test(A, B, alternative="less")
print("2.2.1 A < B :", p1, "| medians:", medA1, medB1)


# ================================================================
# 📘 2.2.2 median₁ > median₂
# ================================================================
premium = [1300, 1500, 1700, 1800]
basic   = [900, 1100, 950, 1050]

p2, medP2, medB2 = mw_test(premium, basic, alternative="greater")
print("2.2.2 premium > basic :", p2, "| medians:", medP2, medB2)


# ================================================================
# 📘 2.2.3 median₁ ≠ median₂
# ================================================================
city1 = [33, 35, 37, 40]
city2 = [45, 47, 42, 50]

p3, m13, m23 = mw_test(city1, city2, alternative="two-sided")
print("2.2.3 city1 != city2 :", p3, "| medians:", m13, m23)


# ================================================================
# 📘 2.2.4 Разные распределения
# ================================================================
exp1 = np.random.exponential(scale=5, size=200)
exp2 = np.random.exponential(scale=9, size=200)

p4, e1, e2 = mw_test(exp1, exp2)
print("2.2.4 exp distributions :", p4, "| medians:", e1, e2)


# ================================================================
# 📘 2.2.5 Выбросы
# ================================================================
clean = [10, 11, 9, 10, 12, 11]
messy = [10, 12, 300, 9, 8, 500]

p5, c1, c2 = mw_test(clean, messy)
print("2.2.5 outliers case :", p5, "| medians:", c1, c2)


# ================================================================
# 📘 2.2.6 Малый объём (n = 5)
# ================================================================
small1 = [1, 2, 3, 4, 5]
small2 = [10, 9, 8, 7, 6]

p6, s1, s2 = mw_test(small1, small2)
print("2.2.6 small sample :", p6, "| medians:", s1, s2)


# ================================================================
# 📘 2.2.7 Бизнес-кейс — скорость обслуживания
# ================================================================
# Проверить: отдел A обслуживает быстрее?
teamA = [120, 110, 115, 130, 125]
teamB = [150, 160, 140, 170, 155]

p7, tA, tB = mw_test(teamA, teamB, alternative="less")
print("2.2.7 teamA faster than B :", p7, "| medians:", tA, tB)


# ================================================================
# 📘 2.2.8 Бизнес-кейс — клиенты тратят больше?
# ================================================================
store1 = [54000, 56000, 59000, 58000]
store2 = [50000, 51000, 52000, 51500]

p8, s1m, s2m = mw_test(store1, store2, alternative="greater")
print("2.2.8 store1 > store2 :", p8, "| medians:", s1m, s2m)


# ================================================================
# 📘 ИТОГ:
# ------------------------------------------------
# ✔ median₁ < median₂                   (2.2.1)
# ✔ median₁ > median₂                   (2.2.2)
# ✔ median₁ ≠ median₂                   (2.2.3)
# ✔ разные распределения                (2.2.4)
# ✔ выбросы                             (2.2.5)
# ✔ малый объём                         (2.2.6)
# ✔ бизнес-кейсы                        (2.2.7–8)
# ------------------------------------------------
# Mann–Whitney часто лучший выбор, когда данные НЕ нормальные.
# Но его ограничение:
#   — предполагает одинаковую форму распределений!
#
# Если это не так → лучше использовать Brunner–Munzel (см. 2.1.d).
# ================================================================
```

```python
# ================================================================
# 📘 2.1.a ФУНКЦИЯ: Welch t-test (независимые выборки)
# ================================================================
# Используется когда:
#   • сравниваем две независимые группы
#   • нормальность *примерно* выполняется (рекомендовано)
#   • дисперсии РАЗНЫЕ (главное преимущество Welch)
#   • размеры групп могут быть разными
#
# Проверяет:
#   H0: mean(g1) = mean(g2)
#   H1: mean(g1) < mean(g2)
#   H1: mean(g1) > mean(g2)
#   H1: mean(g1) ≠ mean(g2)
#
# Вызов:
#   stats.ttest_ind(g1, g2, equal_var=False, alternative="...")
# ================================================================

import numpy as np
from scipy.stats import ttest_ind


def welch_test(g1, g2, alternative="two-sided"):
    """
    Wrapper над ttest_ind(equal_var=False)
    """
    result = ttest_ind(g1, g2, equal_var=False, alternative=alternative)
    return result.statistic, result.pvalue, np.mean(g1), np.mean(g2)


# ================================================================
# 📘 2.1.a.1 mean₁ < mean₂
# ================================================================
# Пример: средняя скорость отдела A < отдела B?
A = [10, 12, 11, 9, 10]
B = [15, 14, 16, 17, 13]

stat1, p1, mA1, mB1 = welch_test(A, B, alternative="less")
print("2.1.a.1 A < B:", p1, "| means:", mA1, mB1)


# ================================================================
# 📘 2.1.a.2 mean₁ > mean₂
# ================================================================
premium = [1200, 1500, 1700, 1800]
basic   = [900, 1000, 1100, 1050]

stat2, p2, m1, m2 = welch_test(premium, basic, alternative="greater")
print("2.1.a.2 premium > basic:", p2, "| means:", m1, m2)


# ================================================================
# 📘 2.1.a.3 mean₁ ≠ mean₂
# ================================================================
north = [33, 35, 37, 40]
south = [45, 47, 42, 50]

stat3, p3, m13, m23 = welch_test(north, south, alternative="two-sided")
print("2.1.a.3 north != south:", p3, "| means:", m13, m23)


# ================================================================
# 📘 2.1.a.4 Разные дисперсии (главный случай Welch)
# ================================================================
low_var  = np.random.normal(50, 1, 200)   # маленькая дисперсия
high_var = np.random.normal(52, 10, 200)  # большая дисперсия

stat4, p4, lv, hv = welch_test(low_var, high_var)
print("2.1.a.4 unequal variances:", p4, "| means:", lv, hv)


# ================================================================
# 📘 2.1.a.5 Выбросы (Welch лучше классического t-test)
# ================================================================
clean = [10, 11, 12, 10, 9]
messy = [10, 11, 300, 9, 8]

stat5, p5, c1, c2 = welch_test(clean, messy)
print("2.1.a.5 with outliers:", p5, "| means:", c1, c2)


# ================================================================
# 📘 2.1.a.6 Малый объём данных
# ================================================================
g_small1 = [1, 2, 3, 4]
g_small2 = [10, 9, 8, 7]

stat6, p6, s1, s2 = welch_test(g_small1, g_small2)
print("2.1.a.6 small samples:", p6, "| means:", s1, s2)


# ================================================================
# 📘 2.1.a.7 Бизнес-кейс — время обработки заявок
# ================================================================
old = [40, 42, 41, 39, 38]
new = [35, 34, 32, 33, 36]

stat7, p7, o, n = welch_test(old, new, alternative="greater")
print("2.1.a.7 old > new:", p7, "| means:", o, n)


# ================================================================
# 📘 2.1.a.8 Бизнес-кейс — сравнение двух магазинов по выручке
# ================================================================
shop1 = [54000, 56000, 59000, 58000]
shop2 = [50000, 51000, 52000, 51500]

stat8, p8, s1m, s2m = welch_test(shop1, shop2, alternative="greater")
print("2.1.a.8 shop1 > shop2:", p8, "| means:", s1m, s2m)


# ================================================================
# 📘 ИТОГ:
# ------------------------------------------------
# ✔ mean₁ < mean₂                     (2.1.a.1)
# ✔ mean₁ > mean₂                     (2.1.a.2)
# ✔ mean₁ ≠ mean₂                     (2.1.a.3)
# ✔ разные дисперсии                  (2.1.a.4)
# ✔ выбросы                           (2.1.a.5)
# ✔ малые выборки                     (2.1.a.6)
# ✔ бизнес-кейсы                      (2.1.a.7–8)
# ------------------------------------------------
# Welch t-test лучше классического t-test когда:
#   1) дисперсии отличаются
#   2) размеры выборок разные
#   3) данные слегка не нормальные
# ================================================================
```

```python
# ================================================================
# 📘 2.1.b ФУНКЦИЯ: Permutation 2-sample test
# ================================================================
# Используется когда:
#   • сравниваем две независимые группы
#   • НЕТ нормальности
#   • НЕТ равенства дисперсий
#   • сильно разные формы распределения
#   • мало данных
#   • присутствуют выбросы
#
# Это один из самых надёжных тестов.
#
# Проверяет:
#   H0: mean(g1) = mean(g2)
#
# Параметры:
#   alternative: 'two-sided', 'greater', 'less'
#   n_perm — число перестановок (10000 оптимально)
# ================================================================

import numpy as np


def permutation_test(g1, g2, alternative="two-sided", n_perm=10000):
    g1 = np.asarray(g1)
    g2 = np.asarray(g2)

    observed = np.mean(g1) - np.mean(g2)
    combined = np.concatenate([g1, g2])
    n1 = len(g1)

    # пермутации
    perms = []
    for _ in range(n_perm):
        np.random.shuffle(combined)
        new_g1 = combined[:n1]
        new_g2 = combined[n1:]
        perms.append(np.mean(new_g1) - np.mean(new_g2))

    perms = np.array(perms)

    if alternative == "two-sided":
        p = np.mean(np.abs(perms) >= abs(observed))
    elif alternative == "greater":
        p = np.mean(perms >= observed)
    else:  # "less"
        p = np.mean(perms <= observed)

    return observed, p, np.mean(g1), np.mean(g2)


# ================================================================
# 📘 2.1.b.1 mean₁ < mean₂
# ================================================================
A = [41, 42, 40, 39]
B = [50, 48, 49, 52]

obs1, p1, mA1, mB1 = permutation_test(A, B, alternative="less")
print("2.1.b.1 A < B :", p1, "| means:", mA1, mB1)


# ================================================================
# 📘 2.1.b.2 mean₁ > mean₂
# ================================================================
premium = [1200, 1500, 1700, 1800]
basic   = [900, 1000, 1100, 1050]

obs2, p2, mp, mb = permutation_test(premium, basic, alternative="greater")
print("2.1.b.2 premium > basic :", p2, "| means:", mp, mb)


# ================================================================
# 📘 2.1.b.3 mean₁ ≠ mean₂
# ================================================================
north = [33, 35, 37, 40]
south = [45, 47, 42, 50]

obs3, p3, m1, m2 = permutation_test(north, south, alternative="two-sided")
print("2.1.b.3 north != south :", p3, "| means:", m1, m2)


# ================================================================
# 📘 2.1.b.4 Разные формы распределений
# ================================================================
exp1 = np.random.exponential(scale=5, size=150)
exp2 = np.random.normal(loc=10, scale=5, size=150)

obs4, p4, e1, e2 = permutation_test(exp1, exp2)
print("2.1.b.4 different shapes :", p4)


# ================================================================
# 📘 2.1.b.5 Выбросы (пермутации устойчивы)
# ================================================================
clean = [10, 11, 12, 10, 9]
messy = [10, 12, 300, 9, 8]

obs5, p5, c1, c2 = permutation_test(clean, messy)
print("2.1.b.5 outliers :", p5)


# ================================================================
# 📘 2.1.b.6 Малый объём данных
# ================================================================
g_small1 = [1, 2, 3]
g_small2 = [10, 9, 8]

obs6, p6, s1, s2 = permutation_test(g_small1, g_small2)
print("2.1.b.6 small sample :", p6)


# ================================================================
# 📘 2.1.b.7 Большие различия дисперсий
# ================================================================
low_var  = np.random.normal(20, 1, 200)
high_var = np.random.normal(24, 10, 200)

obs7, p7, lv, hv = permutation_test(low_var, high_var)
print("2.1.b.7 unequal variances :", p7)


# ================================================================
# 📘 2.1.b.8 Бизнес-кейс — интерфейс ускорил работу?
# ================================================================
old = [40, 42, 41, 39, 38, 37]
new = [33, 34, 35, 36, 32, 30]

obs8, p8, o, n = permutation_test(old, new, alternative="greater")
print("2.1.b.8 old > new :", p8, "| means:", o, n)


# ================================================================
# 📘 2.1.b.9 Бизнес-кейс — разница конверсии двух сайтов
# ================================================================
siteA = np.random.beta(2, 5, size=300) * 100  # %
siteB = np.random.beta(4, 6, size=300) * 100

obs9, p9, sa, sb = permutation_test(siteA, siteB)
print("2.1.b.9 conversion:", p9)


# ================================================================
# 📘 ИТОГ:
# ------------------------------------------------
# ✔ mean₁ < mean₂                    (2.1.b.1)
# ✔ mean₁ > mean₂                    (2.1.b.2)
# ✔ mean₁ ≠ mean₂                    (2.1.b.3)
# ✔ разные формы распределений       (2.1.b.4)
# ✔ устойчив к выбросам              (2.1.b.5)
# ✔ работает при малом n             (2.1.b.6)
# ✔ дисперсии могут быть разными     (2.1.b.7)
# ✔ бизнес-кейсы                     (2.1.b.8–9)
# ------------------------------------------------
# Permutation test — один из самых мощных тестов:
#   • полностью непараметрический
#   • даёт точный p-value
#   • не требует нормальности
#   • отлично работает при выбросах
#   • не требует равенства дисперсий
# ================================================================
```

```python
# ================================================================
# 📘 2.1.c ФУНКЦИЯ: Kolmogorov–Smirnov 2-sample test
# ================================================================
# Используется когда:
#   • сравниваем две независимые выборки
#   • НЕ нормальные данные
#   • подозрение, что распределения разные
#   • разница в форме/вариации/хвостах
#
# Проверяет:
#   H0: оба распределения одинаковы
#   H1: распределения различаются (двусторонний KS)
#
# SciPy:
#   stats.ks_2samp(g1, g2, alternative="two-sided")
# ================================================================

import numpy as np
from scipy.stats import ks_2samp


def ks_test(g1, g2, alternative="two-sided"):
    result = ks_2samp(g1, g2, alternative=alternative)
    return result.statistic, result.pvalue


# ================================================================
# 📘 2.1.c.1 Распределения отличаются (основной случай)
# ================================================================
A = np.random.exponential(scale=5, size=200)
B = np.random.normal(loc=10, scale=5, size=200)

stat1, p1 = ks_test(A, B)
print("2.1.c.1 distributions differ:", p1)


# ================================================================
# 📘 2.1.c.2 mean₁ < mean₂ (через CDF)
# ================================================================
# В KS-test нет 'less/greater' для средних, но можно использовать это как:
#   H1: CDF_g1(x) > CDF_g2(x) → g1 < g2
# ================================================================
low = np.random.normal(30, 5, 200)
high = np.random.normal(50, 5, 200)

stat2, p2 = ks_test(low, high, alternative="less")
print("2.1.c.2 g1 < g2 (CDF test):", p2)


# ================================================================
# 📘 2.1.c.3 mean₁ > mean₂
# ================================================================
stat3, p3 = ks_test(low, high, alternative="greater")
print("2.1.c.3 g1 > g2 (CDF test):", p3)


# ================================================================
# 📘 2.1.c.4 Выбросы (KS устойчив)
# ================================================================
clean = [10, 11, 9, 10, 12, 11, 8, 9]
messy = [10, 12, 300, 9, 8, 500, 400]

stat4, p4 = ks_test(clean, messy)
print("2.1.c.4 outliers:", p4)


# ================================================================
# 📘 2.1.c.5 Очень разные формы распределений
# ================================================================
uni = np.random.uniform(0, 1, 300)
beta = np.random.beta(2, 5, 300)

stat5, p5 = ks_test(uni, beta)
print("2.1.c.5 different shapes:", p5)


# ================================================================
# 📘 2.1.c.6 Разные дисперсии
# ================================================================
n1 = np.random.normal(10, 1, 300)
n2 = np.random.normal(10, 10, 300)

stat6, p6 = ks_test(n1, n2)
print("2.1.c.6 unequal variances:", p6)


# ================================================================
# 📘 2.1.c.7 Малый объём выборок
# ================================================================
small1 = [1, 2, 3, 4]
small2 = [10, 9, 8, 7]

stat7, p7 = ks_test(small1, small2)
print("2.1.c.7 small samples:", p7)


# ================================================================
# 📘 2.1.c.8 Бизнес-кейс — изменения поведения клиентов
# ================================================================
before = np.random.gamma(shape=2, scale=20, size=400)
after  = np.random.gamma(shape=4, scale=12, size=400)

stat8, p8 = ks_test(before, after)
print("2.1.c.8 customers distribution changed:", p8)


# ================================================================
# 📘 2.1.c.9 Бизнес-кейс — сравнение распределений продаж
# ================================================================
storeA = np.random.lognormal(mean=10, sigma=0.3, size=300)
storeB = np.random.lognormal(mean=10, sigma=0.7, size=300)

stat9, p9 = ks_test(storeA, storeB)
print("2.1.c.9 sales distribution difference:", p9)


# ================================================================
# 📘 ИТОГ:
# ------------------------------------------------
# ✔ различие распределений                (2.1.c.1)
# ✔ проверка г1 < г2 через CDF            (2.1.c.2)
# ✔ проверка г1 > г2                      (2.1.c.3)
# ✔ устойчивость к выбросам               (2.1.c.4)
# ✔ разные формы распределений            (2.1.c.5)
# ✔ разные дисперсии                      (2.1.c.6)
# ✔ малые выборки                         (2.1.c.7)
# ✔ бизнес-кейсы                          (2.1.c.8–9)
# ------------------------------------------------
# KS-test — лучший тест для сравнения распределений полностью,
# а не только среднего или медианы.
# ================================================================
```

```python
# ================================================================
# 📘 2.1.d ФУНКЦИЯ: Brunner–Munzel 2-sample test (ручная реализация)
# ================================================================
# Используется когда:
#   • сравниваем две независимые группы
#   • данные НЕ нормальные
#   • дисперсии НЕ равны
#   • формы распределений НЕ одинаковые
#   • Mann–Whitney U НЕ подходит
#
# Проверяет:
#   H0: распределения равны → P(X > Y) = 0.5
#   H1: P(X > Y) ≠ 0.5        (двусторонний)
#   или H1: > / <
# ================================================================

import numpy as np
from scipy.stats import t


def brunner_munzel_test(x, y, alternative="two-sided"):
    """
    Реализация Brunner–Munzel test без statsmodels.
    Работает на Python 3.12–3.13.
    """
    x = np.asarray(x)
    y = np.asarray(y)
    nx = len(x)
    ny = len(y)

    # ранги объединённых данных
    ranks = np.argsort(np.argsort(np.concatenate([x, y]))) + 1
    rx = ranks[:nx]
    ry = ranks[nx:]

    # средние ранги
    mean_rx = np.mean(rx)
    mean_ry = np.mean(ry)

    # параметр W (оценка вероятности превосходства)
    w = (mean_rx - mean_ry) / (nx + ny)

    # дисперсии рангов
    sx = np.sum((rx - mean_rx)**2) / (nx - 1)
    sy = np.sum((ry - mean_ry)**2) / (ny - 1)

    se = np.sqrt(sx / nx + sy / ny)  # стандартная ошибка
    t_stat = w / se                  # t статистика

    # степени свободы (формула Уэлча)
    df = (sx / nx + sy / ny)**2 / (
        (sx**2) / (nx**2 * (nx - 1)) +
        (sy**2) / (ny**2 * (ny - 1))
    )

    # p-value
    if alternative == "two-sided":
        p = 2 * (1 - t.cdf(abs(t_stat), df))
    elif alternative == "greater":
        p = 1 - t.cdf(t_stat, df)
    else:  # "less"
        p = t.cdf(t_stat, df)

    return t_stat, df, p


# удобное короткое имя
bm_test = brunner_munzel_test


# ================================================================
# 📘 2.1.d.1 median₁ < median₂
# ================================================================
A = [40, 42, 38, 41, 37]
B = [47, 49, 45, 50, 52]

t1, df1, p1 = bm_test(A, B, alternative="less")
print("2.1.d.1 A < B:", p1)


# ================================================================
# 📘 2.1.d.2 median₁ > median₂
# ================================================================
premium = [1200, 1500, 1700, 1800]
basic   = [900, 1000, 1100, 1050]

t2, df2, p2 = bm_test(premium, basic, alternative="greater")
print("2.1.d.2 premium > basic:", p2)


# ================================================================
# 📘 2.1.d.3 median₁ ≠ median₂
# ================================================================
city1 = [33, 35, 37, 40]
city2 = [45, 47, 42, 50]

t3, df3, p3 = bm_test(city1, city2, alternative="two-sided")
print("2.1.d.3 city1 != city2:", p3)


# ================================================================
# 📘 2.1.d.4 Разные формы распределений
# ================================================================
expo1 = np.random.exponential(scale=3, size=200)
expo2 = np.random.exponential(scale=7, size=200)

t4, df4, p4 = bm_test(expo1, expo2)
print("2.1.d.4 different distribution shapes:", p4)


# ================================================================
# 📘 2.1.d.5 Выбросы
# ================================================================
clean = [10, 11, 9, 10, 12, 11]
messy = [10, 12, 300, 9, 8, 500]  # выбросы

t5, df5, p5 = bm_test(clean, messy)
print("2.1.d.5 outliers:", p5)


# ================================================================
# 📘 2.1.d.6 Малый объём данных
# ================================================================
g_small1 = [1, 2, 3, 4]
g_small2 = [10, 9, 8, 7]

t6, df6, p6 = bm_test(g_small1, g_small2)
print("2.1.d.6 small sample:", p6)


# ================================================================
# 📘 2.1.d.7 Большие различия дисперсий
# ================================================================
low_var  = np.random.normal(10, 1, 200)
high_var = np.random.normal(10, 10, 200)

t7, df7, p7 = bm_test(low_var, high_var)
print("2.1.d.7 unequal variances:", p7)


# ================================================================
# 📘 2.1.d.8 Бизнес-кейс — сравнение поведения клиентов
# ================================================================
before = np.random.gamma(shape=2, scale=20, size=400)
after  = np.random.gamma(shape=5, scale=10, size=400)

t8, df8, p8 = bm_test(before, after)
print("2.1.d.8 customers changed:", p8)


# ================================================================
# 📘 2.1.d.9 Бизнес-кейс — скорость обработки задач
# ================================================================
team_A = np.random.exponential(scale=10, size=300)
team_B = np.random.exponential(scale=5, size=300)

t9, df9, p9 = bm_test(team_A, team_B, alternative="greater")
print("2.1.d.9 team_A slower than B:", p9)


# ================================================================
# 📘 ИТОГ:
# ------------------------------------------------
# ✔ median₁ < median₂                     (2.1.d.1)
# ✔ median₁ > median₂                     (2.1.d.2)
# ✔ median₁ ≠ median₂                     (2.1.d.3)
# ✔ разные формы распределений            (2.1.d.4)
# ✔ выбросы                               (2.1.d.5)
# ✔ малые выборки                         (2.1.d.6)
# ✔ разные дисперсии                      (2.1.d.7)
# ✔ бизнес-кейсы                          (2.1.d.8–9)
# ------------------------------------------------
# Brunner–Munzel лучше Mann–Whitney когда:
#   1) дисперсии различаются
#   2) распределения имеют разную форму
#   3) наблюдения не "пересекаются" по рангам
#   4) выборки сильно отличаются по масштабу
# ================================================================
```

```python
# ================================================================
# 📘 2.2 ФУНКЦИЯ: Mann–Whitney U-test (две независимые группы)
# ================================================================
# Проверяет:
#   H0: распределения одинаковы
#   H1: медианы различаются (≠) или < или >
#
# Параметры:
#   alternative: 'two-sided', 'less', 'greater'
# ================================================================

import numpy as np
from scipy.stats import mannwhitneyu


def mw_test(g1, g2, alternative="two-sided"):
    """
    Оболочка для mannwhitneyu с возвратом медиан.
    """
    g1 = np.asarray(g1)
    g2 = np.asarray(g2)

    res = mannwhitneyu(g1, g2, alternative=alternative)
    p = res.pvalue

    return p, np.median(g1), np.median(g2)


# ================================================================
# 📘 2.2.1 median₁ < median₂
# ================================================================
# Пример: производительность отдела A < отдела B?
A = [40, 42, 38, 41, 37]
B = [55, 52, 49, 50, 53]

p1, mA1, mB1 = mw_test(A, B, alternative="less")
print("2.2.1 A < B:", p1, "| medians:", mA1, mB1)


# ================================================================
# 📘 2.2.2 median₁ > median₂
# ================================================================
premium = [1500, 1700, 1800, 1600]
basic   = [900, 1000, 1050, 1100]

p2, mP2, mB2 = mw_test(premium, basic, alternative="greater")
print("2.2.2 premium > basic:", p2, "| medians:", mP2, mB2)


# ================================================================
# 📘 2.2.3 median₁ ≠ median₂
# ================================================================
north = [33, 35, 37, 40]
south = [48, 45, 50, 47]

p3, mN, mS = mw_test(north, south, alternative="two-sided")
print("2.2.3 north != south:", p3, "| medians:", mN, mS)


# ================================================================
# 📘 2.2.4 Разные формы распределений (MW работает, но BM лучше)
# ================================================================
exp1 = np.random.exponential(scale=5, size=200)
exp2 = np.random.exponential(scale=8, size=200)

p4, me1, me2 = mw_test(exp1, exp2)
print("2.2.4 different shapes:", p4, "| medians:", me1, me2)


# ================================================================
# 📘 2.2.5 Выбросы
# ================================================================
clean = [10, 11, 9, 10, 12]
messy = [10, 12, 300, 9, 500]

p5, c1, c2 = mw_test(clean, messy)
print("2.2.5 outliers:", p5, "| medians:", c1, c2)


# ================================================================
# 📘 2.2.6 Малый объём выборок
# ================================================================
small1 = [1, 2, 3]
small2 = [10, 12, 9]

p6, s1, s2 = mw_test(small1, small2)
print("2.2.6 small samples:", p6, "| medians:", s1, s2)


# ================================================================
# 📘 2.2.7 Бизнес-кейс — скорость обработки заявок
# ================================================================
old = [40, 45, 42, 44]
new = [30, 32, 35, 33]

p7, o1, o2 = mw_test(old, new, alternative="greater")
print("2.2.7 old > new (slower):", p7, "| medians:", o1, o2)


# ================================================================
# 📘 2.2.8 Бизнес-кейс — разница в тратах клиентов
# ================================================================
storeA = [54000, 56000, 59000, 60000]
storeB = [50000, 51000, 52000, 51500]

p8, sA, sB = mw_test(storeA, storeB, alternative="greater")
print("2.2.8 storeA > storeB:", p8, "| medians:", sA, sB)


# ================================================================
# 📘 ИТОГ:
# ------------------------------------------------
# ✔ median₁ < median₂                   (2.2.1)
# ✔ median₁ > median₂                   (2.2.2)
# ✔ median₁ ≠ median₂                   (2.2.3)
# ✔ устойчив к выбросам                 (2.2.5)
# ✔ работает при малом n                (2.2.6)
# ✔ бизнес-кейсы                        (2.2.7–8)
# ------------------------------------------------
# Mann–Whitney хорош когда:
#   • данные НЕ нормальные
#   • выбросы присутствуют
#   • выборки независимые
#   • хочется сравнить медианы
#
# Если формы распределений НЕ одинаковые → использовать Brunner–Munzel (2.1.d)
# ================================================================
```

```python
# ================================================================
# 📘 2.2.a Brunner–Munzel test (ALTERNATIVE to Mann–Whitney)
# ================================================================
# Использовать вместо MWU, когда:
#   • дисперсии разные
#   • формы распределений разные
#   • MW нарушает предположения
#   • хотим оценить P(X > Y)
#
# Проверяет:
#   H0: распределения равны → P(X > Y) = 0.5
#   H1: P(X > Y) ≠ 0.5
#
# Функция — ручная реализация (нет в statsmodels под Python 3.12+)
# ================================================================

import numpy as np
from scipy.stats import t


def brunner_munzel_test(x, y, alternative="two-sided"):
    """
    Реализация Brunner–Munzel test без statsmodels.
    Работает на Python 3.12–3.13.
    """
    x = np.asarray(x)
    y = np.asarray(y)
    nx = len(x)
    ny = len(y)

    # ранги объединённых данных
    ranks = np.argsort(np.argsort(np.concatenate([x, y]))) + 1
    rx = ranks[:nx]
    ry = ranks[nx:]

    # средние ранги
    mean_rx = np.mean(rx)
    mean_ry = np.mean(ry)

    # параметр W
    w = (mean_rx - mean_ry) / (nx + ny)

    # дисперсии рангов
    sx = np.sum((rx - mean_rx)**2) / (nx - 1)
    sy = np.sum((ry - mean_ry)**2) / (ny - 1)

    se = np.sqrt(sx / nx + sy / ny)  # standard error
    t_stat = w / se

    # степени свободы (Welch formula)
    df = (sx / nx + sy / ny)**2 / (
        (sx**2) / (nx**2 * (nx - 1)) +
        (sy**2) / (ny**2 * (ny - 1))
    )

    # p-value
    if alternative == "two-sided":
        p = 2 * (1 - t.cdf(abs(t_stat), df))
    elif alternative == "greater":
        p = 1 - t.cdf(t_stat, df)
    else:  # "less"
        p = t.cdf(t_stat, df)

    return t_stat, df, p


# короткое имя — так проще вызывать
bm = brunner_munzel_test


# ================================================================
# 📘 2.2.a.1 median₁ < median₂
# ================================================================
A = [40, 42, 38, 41, 37]
B = [55, 52, 49, 50, 53]

t1, df1, p1 = bm(A, B, alternative="less")
print("2.2.a.1 A < B:", p1)


# ================================================================
# 📘 2.2.a.2 median₁ > median₂
# ================================================================
premium = [1500, 1700, 1800, 1600]
basic   = [900, 1000, 1050, 1100]

t2, df2, p2 = bm(premium, basic, alternative="greater")
print("2.2.a.2 premium > basic:", p2)


# ================================================================
# 📘 2.2.a.3 median₁ ≠ median₂
# ================================================================
north = [33, 35, 37, 40]
south = [48, 45, 50, 47]

t3, df3, p3 = bm(north, south, alternative="two-sided")
print("2.2.a.3 north != south:", p3)


# ================================================================
# 📘 2.2.a.4 Разные формы распределений
# ================================================================
exp1 = np.random.exponential(scale=5, size=200)
exp2 = np.random.uniform(0, 50, size=200)

t4, df4, p4 = bm(exp1, exp2)
print("2.2.a.4 different shapes:", p4)


# ================================================================
# 📘 2.2.a.5 Выбросы
# ================================================================
clean = [10, 11, 9, 10, 12]
messy = [10, 12, 300, 9, 500]

t5, df5, p5 = bm(clean, messy)
print("2.2.a.5 outliers:", p5)


# ================================================================
# 📘 2.2.a.6 Малые выборки
# ================================================================
s1 = [1, 2, 3]
s2 = [10, 12, 9]

t6, df6, p6 = bm(s1, s2)
print("2.2.a.6 small samples:", p6)


# ================================================================
# 📘 2.2.a.7 Большие различия дисперсий
# ================================================================
low_var = np.random.normal(20, 1, 200)
high_var = np.random.normal(20, 12, 200)

t7, df7, p7 = bm(low_var, high_var)
print("2.2.a.7 unequal variances:", p7)


# ================================================================
# 📘 2.2.a.8 Бизнес-кейс — изменение поведения клиентов
# ================================================================
before = np.random.gamma(shape=2, scale=20, size=400)
after  = np.random.gamma(shape=4, scale=12, size=400)

t8, df8, p8 = bm(before, after)
print("2.2.a.8 customers changed:", p8)


# ================================================================
# 📘 Итог:
# ------------------------------------------------
# ✔ median₁ < median₂                    (2.2.a.1)
# ✔ median₁ > median₂                    (2.2.a.2)
# ✔ median₁ ≠ median₂                    (2.2.a.3)
# ✔ разные формы распределений           (2.2.a.4)
# ✔ устойчив к выбросам                  (2.2.a.5)
# ✔ работает при малом n                 (2.2.a.6)
# ✔ выдерживает разные дисперсии         (2.2.a.7)
# ✔ подходит для бизнес-задач            (2.2.a.8)
# ------------------------------------------------
# Brunner–Munzel — лучший непараметрический тест,
# когда Mann–Whitney даёт неверный p-value.
# ================================================================
```

```python
# ================================================================
# 📘 2.2.b Kolmogorov–Smirnov 2-sample test (альтернатива MWU)
# ================================================================
# Использовать когда:
#   • два независимых распределения
#   • НЕ нормальные данные
#   • формы распределений могут различаться
#   • дисперсии могут различаться
#   • нужно сравнивать РАСПРЕДЕЛЕНИЯ полностью
#
# Проверяет:
#   H0: распределения одинаковые
#   H1: распределения отличаются (двусторонний)
#   или направленные: < или >
# ================================================================

import numpy as np
from scipy.stats import ks_2samp


def ks_test(g1, g2, alternative="two-sided"):
    """
    Оболочка вокруг ks_2samp.
    Возвращает D-статистику и p-value.
    """
    g1 = np.asarray(g1)
    g2 = np.asarray(g2)

    res = ks_2samp(g1, g2, alternative=alternative)
    return res.statistic, res.pvalue


# ================================================================
# 📘 2.2.b.1 Распределения ≠ (основной случай KS)
# ================================================================
A = np.random.exponential(scale=5, size=300)
B = np.random.normal(loc=10, scale=5, size=300)

D1, p1 = ks_test(A, B)
print("2.2.b.1 distributions differ:", p1)


# ================================================================
# 📘 2.2.b.2 Проверка г1 < г2 через CDF
# ================================================================
# mean₁ < mean₂ → CDF₁(x) > CDF₂(x)
low  = np.random.normal(30, 5, size=300)
high = np.random.normal(50, 5, size=300)

D2, p2 = ks_test(low, high, alternative="less")
print("2.2.b.2 g1 < g2 (CDF test):", p2)


# ================================================================
# 📘 2.2.b.3 Проверка г1 > г2
# ================================================================
D3, p3 = ks_test(low, high, alternative="greater")
print("2.2.b.3 g1 > g2 (CDF test):", p3)


# ================================================================
# 📘 2.2.b.4 Большие различия форм распределений
# ================================================================
uni  = np.random.uniform(0, 1, size=400)
beta = np.random.beta(2, 5, size=400)

D4, p4 = ks_test(uni, beta)
print("2.2.b.4 very different shapes:", p4)


# ================================================================
# 📘 2.2.b.5 Выбросы (KS устойчевее t-test, но чувствительнее MWU)
# ================================================================
clean = [10, 11, 9, 10, 12]
messy = [10, 12, 200, 300, 8, 500]

D5, p5 = ks_test(clean, messy)
print("2.2.b.5 outliers effect:", p5)


# ================================================================
# 📘 2.2.b.6 Разные дисперсии
# ================================================================
n1 = np.random.normal(10, 1, 400)
n2 = np.random.normal(10, 12, 400)

D6, p6 = ks_test(n1, n2)
print("2.2.b.6 unequal variances:", p6)


# ================================================================
# 📘 2.2.b.7 Малый объём данных
# ================================================================
small1 = [1, 2, 3, 4]
small2 = [10, 9, 8, 7]

D7, p7 = ks_test(small1, small2)
print("2.2.b.7 small samples:", p7)


# ================================================================
# 📘 2.2.b.8 Бизнес-кейс — изменение распределения времени доставки
# ================================================================
before = np.random.gamma(2, 20, 400)
after  = np.random.gamma(4, 12, 400)

D8, p8 = ks_test(before, after)
print("2.2.b.8 distribution changed:", p8)


# ================================================================
# 📘 2.2.b.9 Бизнес-кейс — распределения продаж по магазинам
# ================================================================
shopA = np.random.lognormal(mean=10, sigma=0.3, size=500)
shopB = np.random.lognormal(mean=10, sigma=0.7, size=500)

D9, p9 = ks_test(shopA, shopB)
print("2.2.b.9 sales distributions differ:", p9)


# ================================================================
# 📘 ИТОГ:
# ------------------------------------------------
# ✔ сравнение распределений полностью      (2.2.b.1)
# ✔ направленные гипотезы через CDF        (2.2.b.2–3)
# ✔ очень разные формы распределений       (2.2.b.4)
# ✔ устойчив к выбросам                    (2.2.b.5)
# ✔ разные дисперсии                       (2.2.b.6)
# ✔ работает при малом n                   (2.2.b.7)
# ✔ бизнес-кейсы                           (2.2.b.8–9)
# ------------------------------------------------
# KS-test — альтернатива Mann–Whitney, когда:
#   • нас интересует форма распределения
#   • MWU не распознаёт различия в хвостах
#   • дисперсии сильно отличаются
#   • распределения имеют разную форму
# ================================================================
```

```python
# ================================================================
# 📘 2.2.c Median test (2-sample) — FULL FIXED VERSION
# ================================================================
# Использовать когда:
#   • сравниваем две независимые группы
#   • данные НЕ нормальные
#   • присутствуют выбросы
#   • интересует ИМЕННО разница медиан
#
# Проверяет:
#   H0: median(g1) = median(g2)
#   H1: медианы различаются (two-sided)
#   H1: median(g1) > median(g2)   (greater)
#   H1: median(g1) < median(g2)   (less)
#
# SciPy median_test не даёт корректную медиану => вычисляем вручную.
# ================================================================

import numpy as np
from scipy.stats import median_test


def median_test_2sample(g1, g2, alternative="two-sided"):
    """
    Исправленный median-test с поддержкой направленных альтернатив.

    alternative:
        'two-sided'
        'greater' → median(g1) > median(g2)
        'less'    → median(g1) < median(g2)

    Возвращает:
        chi2, p-value, grand_median
    """

    g1 = np.asarray(g1)
    g2 = np.asarray(g2)

    # chi² и двусторонний p-value — корректны
    chi2, p, table, _ = median_test(g1, g2)

    # ВАЖНО: вручную считаем общую медиану
    grand_median = np.median(np.concatenate([g1, g2]))

    # Двусторонний тест — готов
    if alternative == "two-sided":
        return chi2, p, grand_median

    # Доли элементов, превышающих медиану
    p1 = np.mean(g1 > grand_median)
    p2 = np.mean(g2 > grand_median)

    # Односторонние p-value
    if alternative == "greater":  # median(g1) > median(g2)
        p_one = p / 2 if p1 > p2 else 1 - p/2
        return chi2, p_one, grand_median

    if alternative == "less":  # median(g1) < median(g2)
        p_one = p / 2 if p1 < p2 else 1 - p/2
        return chi2, p_one, grand_median


# ================================================================
# 📘 2.2.c.1 Медина₁ ≠ Медиана₂ (основной случай)
# ================================================================

A = [10, 12, 14, 11, 13]
B = [20, 22, 25, 21, 23]

chi2_1, p1, med1 = median_test_2sample(A, B)
print("2.2.c.1 median != median :", p1, "| grand median =", med1)


# ================================================================
# 📘 2.2.c.2 median₁ < median₂ (направленная гипотеза)
# ================================================================

low  = [5, 6, 7, 8, 6]
high = [10, 12, 11, 13, 15]

chi2_2, p2, med2 = median_test_2sample(low, high, alternative="less")
print("2.2.c.2 median1 < median2 :", p2, "| grand median =", med2)


# ================================================================
# 📘 2.2.c.3 median₁ > median₂
# ================================================================

premium = [200, 220, 240, 260]
basic   = [100, 120, 140, 160]

chi2_3, p3, med3 = median_test_2sample(premium, basic, alternative="greater")
print("2.2.c.3 median1 > median2 :", p3, "| grand median =", med3)


# ================================================================
# 📘 2.2.c.4 Разные формы распределений (median test устойчив)
# ================================================================

expo1 = np.random.exponential(scale=1, size=300)
expo2 = np.random.exponential(scale=5, size=300)

chi2_4, p4, med4 = median_test_2sample(expo1, expo2)
print("2.2.c.4 diff shapes:", p4, "| grand median =", med4)


# ================================================================
# 📘 2.2.c.5 Выбросы
# ================================================================

clean = [10, 11, 12, 11, 13]
messy = [9, 8, 10, 500, 600]

chi2_5, p5, med5 = median_test_2sample(clean, messy)
print("2.2.c.5 outliers:", p5, "| grand median =", med5)


# ================================================================
# 📘 2.2.c.6 Разные дисперсии
# ================================================================

d1 = np.random.normal(50, 2, size=400)
d2 = np.random.normal(50, 20, size=400)

chi2_6, p6, med6 = median_test_2sample(d1, d2)
print("2.2.c.6 unequal variances:", p6, "| median =", med6)


# ================================================================
# 📘 2.2.c.7 Малый объём данных
# ================================================================

s1 = [1, 2, 3]
s2 = [10, 9, 8]

chi2_7, p7, med7 = median_test_2sample(s1, s2)
print("2.2.c.7 small samples:", p7, "| median =", med7)


# ================================================================
# 📘 2.2.c.8 Бизнес-кейс — изменение медианы времени доставки
# ================================================================

before = np.random.gamma(3, 10, 400)
after  = np.random.gamma(3, 8, 400)

chi2_8, p8, med8 = median_test_2sample(before, after)
print("2.2.c.8 delivery median changed:", p8)


# ================================================================
# 📘 2.2.c.9 Бизнес-кейс — медиана продаж по магазинам
# ================================================================

shopA = np.random.lognormal(mean=3, sigma=0.3, size=400)
shopB = np.random.lognormal(mean=3.3, sigma=0.3, size=400)

chi2_9, p9, med9 = median_test_2sample(shopA, shopB)
print("2.2.c.9 sales median diff:", p9)


# ================================================================
# 📘 ИТОГ:
# ------------------------------------------------
# ✔ сравнение медиан                (2.2.c.1)
# ✔ направленные гипотезы          (2.2.c.2–3)
# ✔ устойчив к разным формам       (2.2.c.4)
# ✔ устойчив к выбросам            (2.2.c.5)
# ✔ работает при разных дисперсиях (2.2.c.6)
# ✔ работает при малом n           (2.2.c.7)
# ✔ бизнес-кейсы                   (2.2.c.8–9)
# ------------------------------------------------
# Median test — альтернатива Mann–Whitney, когда:
#   • важна только медиана
#   • MWU даёт противоречивые выводы при выбросах
#   • распределения сильно «грязные»
# ================================================================
```

```python
# ================================================================
# 📘 3.1 Paired t-test (t-тест для зависимых выборок)
# ================================================================
# Использовать когда:
#   • две зависимые группы (до/после, левый/правый глаз, A/B на тех же людях)
#   • данные нормальные или почти нормальные
#
# Проверяет:
#   H0: mean(after - before) = 0
#   H1: mean(after - before) ≠ 0          (two-sided)
#   H1: mean(after - before) < 0          (less)
#   H1: mean(after - before) > 0          (greater)
#
# Применение:
#   • влияние обучения
#   • влияние лекарства на одних и тех же больных
#   • скорость работы сотрудников до/после нового интерфейса
#   • сравнение двух алгоритмов на одной и той же выборке
# ================================================================

import numpy as np
from scipy.stats import ttest_rel


def paired_ttest(before, after, alternative="two-sided"):
    """
    Обёртка вокруг scipy.stats.ttest_rel для зависимых выборок.

    alternative:
        'two-sided'
        'greater' → after > before
        'less'    → after < before

    Возвращает:
        t_stat, p_value
    """
    before = np.asarray(before)
    after  = np.asarray(after)

    result = ttest_rel(before, after, alternative=alternative)

    return result.statistic, result.pvalue


# ================================================================
# 📘 3.1.1 Базовый кейс — время выполнения уменьшилось
# ================================================================
before1 = [20, 22, 24, 26]
after1  = [18, 19, 20, 21]

t1, p1 = paired_ttest(before1, after1, alternative="less")
print("3.1.1 after < before :", p1)


# ================================================================
# 📘 3.1.2 Проверка улучшения (after > before)
# ================================================================
# Например: новые батареи дают большую дальность пробега

before2 = [100, 120, 110, 115]
after2  = [140, 150, 160, 155]

t2, p2 = paired_ttest(before2, after2, alternative="greater")
print("3.1.2 after > before :", p2)


# ================================================================
# 📘 3.1.3 Двусторонний тест (любой сдвиг)
# ================================================================
# Влияние нового интерфейса неизвестно — могло стать лучше или хуже

before3 = [45, 47, 44, 50, 48]
after3  = [47, 48, 46, 49, 47]

t3, p3 = paired_ttest(before3, after3, alternative="two-sided")
print("3.1.3 two-sided:", p3)


# ================================================================
# 📘 3.1.4 Малые выборки (t-test всё ещё работает)
# ================================================================

before4 = [10, 12, 13]
after4  = [11, 12, 14]

t4, p4 = paired_ttest(before4, after4, alternative="two-sided")
print("3.1.4 small sample:", p4)


# ================================================================
# 📘 3.1.5 Данные почти нормальные (не идеально, но допустимо)
# ================================================================
# Нормальность разностей важнее нормальности самих выборок

before5 = np.random.normal(50, 10, 30)
after5  = before5 + np.random.normal(3, 10, 30)  # улучшение в среднем на 3

t5, p5 = paired_ttest(before5, after5)
print("3.1.5 approx normal:", p5)


# ================================================================
# 📘 3.1.6 Бизнес-кейс: обучение сотрудников
# ================================================================
# До обучения они обрабатывали меньше заявок, после — больше

before6 = [40, 42, 38, 41, 37, 39]
after6  = [45, 50, 40, 47, 42, 41]

t6, p6 = paired_ttest(before6, after6, alternative="greater")
print("3.1.6 training effect greater:", p6)


# ================================================================
# 📘 3.1.7 Бизнес-кейс: снижение времени отклика системы
# ================================================================
response_before = [320, 300, 310, 330, 305]
response_after  = [250, 270, 260, 240, 265]

t7, p7 = paired_ttest(response_before, response_after, alternative="less")
print("3.1.7 faster response after update:", p7)


# ================================================================
# 📘 3.1.8 Алгоритмы A и B на одних и тех же данных
# ================================================================
algo_A = [0.87, 0.88, 0.86, 0.89]
algo_B = [0.90, 0.92, 0.91, 0.93]

t8, p8 = paired_ttest(algo_A, algo_B, alternative="greater")
print("3.1.8 algo_B > algo_A:", p8)


# ================================================================
# 📘 ИТОГ:
# ------------------------------------------------
# ✔ after < before                   (3.1.1)
# ✔ after > before                   (3.1.2)
# ✔ двусторонний тест                (3.1.3)
# ✔ работает при малом n             (3.1.4)
# ✔ допускает умеренное нарушение нормальности (3.1.5)
# ✔ бизнес-кейсы обучение            (3.1.6)
# ✔ бизнес-кейсы IT система          (3.1.7)
# ✔ сравнение алгоритмов             (3.1.8)
# ------------------------------------------------
# Paired t-test — лучший выбор, когда:
#   • есть зависимость (до/после)
#   • важны средние значения
#   • разности примерно нормальны
# ================================================================
```

```python
# ================================================================
# 📘 3.1.a Permutation paired test (перестановочный парный тест)
# ================================================================
# Используют когда:
#   • две зависимые группы (до/после)
#   • разности НЕ нормальные
#   • малый объём данных
#   • присутствуют выбросы
#   • t-test неприменим
#
# Идея:
#   Случайно меняем местами пары (before_i, after_i),
#   создаём распределение нулевой гипотезы и считаем p-value.
#
# Проверяет:
#   H0: mean(after - before) = 0
#   H1: mean(after - before) > 0      (greater)
#   H1: mean(after - before) < 0      (less)
#   H1: mean(after - before) ≠ 0      (two-sided)
# ================================================================

import numpy as np


def paired_permutation_test(before, after, n_perm=10000, alternative="two-sided"):
    """
    Перестановочный тест для зависимых выборок.

    alternative:
        'two-sided'
        'greater'
        'less'

    Возвращает:
        observed_diff, p_value
    """

    before = np.asarray(before)
    after  = np.asarray(after)

    # наблюдаемая разница средних
    observed = np.mean(after - before)

    # массив p-value
    diffs = []

    for _ in range(n_perm):
        # случайная перестановка знаков разности
        signs = np.random.choice([1, -1], size=len(before))
        perm_diff = np.mean(signs * (after - before))
        diffs.append(perm_diff)

    diffs = np.array(diffs)

    # --- вычисление p-value по направлению ---
    if alternative == "two-sided":
        p = np.mean(np.abs(diffs) >= np.abs(observed))

    elif alternative == "greater":
        p = np.mean(diffs >= observed)

    elif alternative == "less":
        p = np.mean(diffs <= observed)

    return observed, p


# ================================================================
# 📘 3.1.a.1 after < before  (улучшение)
# ================================================================
before1 = [20, 22, 24, 26]
after1  = [18, 19, 20, 21]

obs1, p1 = paired_permutation_test(before1, after1, alternative="less")
print("3.1.a.1 after < before :", p1, "| observed diff =", obs1)


# ================================================================
# 📘 3.1.a.2 after > before  (ухудшение или рост показателя)
# ================================================================
before2 = [100, 120, 110, 115]
after2  = [140, 150, 160, 155]

obs2, p2 = paired_permutation_test(before2, after2, alternative="greater")
print("3.1.a.2 after > before :", p2, "| observed diff =", obs2)


# ================================================================
# 📘 3.1.a.3 Двусторонний тест
# ================================================================
before3 = [45, 47, 44, 50, 48]
after3  = [47, 48, 46, 49, 47]

obs3, p3 = paired_permutation_test(before3, after3, alternative="two-sided")
print("3.1.a.3 two-sided:", p3, "| diff =", obs3)


# ================================================================
# 📘 3.1.a.4 Малые выборки
# ================================================================
before4 = [10, 12, 13]
after4  = [11, 12, 14]

obs4, p4 = paired_permutation_test(before4, after4)
print("3.1.a.4 small sample:", p4, "| diff =", obs4)


# ================================================================
# 📘 3.1.a.5 Данные с выбросами
# ================================================================
before5 = [100, 102, 101, 500]
after5  = [90, 95, 92, 480]

obs5, p5 = paired_permutation_test(before5, after5, alternative="less")
print("3.1.a.5 outliers:", p5, "| diff =", obs5)


# ================================================================
# 📘 3.1.a.6 Бизнес-кейс — обучение сотрудников
# ================================================================
before6 = [40, 42, 38, 41, 37, 39]
after6  = [45, 50, 40, 47, 42, 41]

obs6, p6 = paired_permutation_test(before6, after6, alternative="greater")
print("3.1.a.6 training effect:", p6, "| diff =", obs6)


# ================================================================
# 📘 3.1.a.7 Бизнес-кейс — ускорение отклика системы
# ================================================================
response_before = [320, 300, 310, 330, 305]
response_after  = [250, 270, 260, 240, 265]

obs7, p7 = paired_permutation_test(response_before, response_after, alternative="less")
print("3.1.a.7 system faster:", p7, "| diff =", obs7)


# ================================================================
# 📘 ИТОГ:
# ------------------------------------------------
# ✔ аналог paired t-test без предположений о нормальности
# ✔ работает с малыми данными
# ✔ устойчив к выбросам
# ✔ подходит для бизнес-кейсов до/после
# ✔ лучший тест при нарушении нормальности разностей
# ------------------------------------------------
# Paired permutation test — лучший выбор, когда:
#   • данные зависимые
#   • распределение разностей неизвестно или «грязное»
#   • t-test применять нельзя
# ================================================================
```

```python
# ================================================================
# 📘 3.1.b Bayesian paired test (Jeffreys–Zellner–Siow Bayesian t-test)
# ================================================================
# Использовать когда:
#   • две зависимые группы (до/после)
#   • нужно НЕ p-value, а отношение вероятностей H1/H0
#   • интересуют вероятности, а не "да/нет"
#   • мультимодальные или неидеально нормальные данные
#   • малый объём выборки
#
# Возвращает:
#   BF10 — Bayes Factor (evidence for H1 over H0)
#   BF01 — обратный (evidence for H0)
#
# Интерпретация BF10:
#   1–3     → слабые доказательства
#   3–10    → умеренные
#   10–30   → сильные
#   30–100  → очень сильные
#   >100    → экстремальные
#
# H1 варианты:
#   two-sided
#   greater
#   less
# ================================================================

import numpy as np
from scipy.stats import t
from math import gamma, sqrt, pi


def bayes_factor_t(t_stat, n, alternative="two-sided", r=0.707):
    """
    Bayes Factor для t-теста (JZS prior, Rouder et al. 2009).

    t_stat — статистика t
    n      — размер выборки
    r      — scale JZS prior (0.707 как в JASP)
    """

    df = n - 1
    g = r**2
    # числитель и знаменатель формулы BF
    numerator = (1 + t_stat**2 / (df * g)) ** (-(df + 1) / 2)
    denominator = (1 + t_stat**2 / df) ** (-(df + 1) / 2)

    bf10 = numerator / denominator

    # корректировка для направленных гипотез
    if alternative == "two-sided":
        return bf10

    # односторонние: используем половину плотности
    p_h1_two = t.sf(abs(t_stat), df)     # вероятность в хвосте
    p_h1_one = t.sf(t_stat, df)          # направленный хвост

    if alternative == "greater":
        return bf10 * (p_h1_one / p_h1_two)

    if alternative == "less":
        return bf10 * (p_h1_one / p_h1_two)

    return bf10


def bayesian_paired_test(before, after, alternative="two-sided"):
    """
    Байесовский paired t-test.

    Возвращает:
        t_stat, BF10, BF01
    """

    before = np.asarray(before)
    after  = np.asarray(after)

    diff = after - before
    n = len(diff)

    # стандартный t-statistic
    t_stat = np.mean(diff) / (np.std(diff, ddof=1) / np.sqrt(n))

    # Bayes Factor
    bf10 = bayes_factor_t(t_stat, n, alternative=alternative)
    bf01 = 1 / bf10

    return t_stat, bf10, bf01


# ================================================================
# 📘 3.1.b.1 after < before (улучшение)
# ================================================================
before1 = [20, 22, 24, 26]
after1  = [18, 19, 20, 21]

t1, bf10_1, bf01_1 = bayesian_paired_test(before1, after1, alternative="less")
print("3.1.b.1 after < before | t =", t1, "| BF10 =", bf10_1, "| BF01 =", bf01_1)


# ================================================================
# 📘 3.1.b.2 after > before (рост показателя)
# ================================================================
before2 = [100, 120, 110, 115]
after2  = [140, 150, 160, 155]

t2, bf10_2, bf01_2 = bayesian_paired_test(before2, after2, alternative="greater")
print("3.1.b.2 after > before | BF10 =", bf10_2)


# ================================================================
# 📘 3.1.b.3 Двусторонний тест
# ================================================================
before3 = [45, 47, 44, 50, 48]
after3  = [47, 48, 46, 49, 47]

t3, bf10_3, bf01_3 = bayesian_paired_test(before3, after3)
print("3.1.b.3 two-sided | BF10 =", bf10_3)


# ================================================================
# 📘 3.1.b.4 Малые выборки
# ================================================================
before4 = [10, 12, 13]
after4  = [11, 12, 14]

t4, bf10_4, bf01_4 = bayesian_paired_test(before4, after4)
print("3.1.b.4 small sample | BF10 =", bf10_4)


# ================================================================
# 📘 3.1.b.5 Данные с выбросами
# ================================================================
before5 = [100, 102, 101, 500]
after5  = [90, 95, 92, 480]

t5, bf10_5, bf01_5 = bayesian_paired_test(before5, after5, alternative="less")
print("3.1.b.5 outliers | BF10 =", bf10_5)


# ================================================================
# 📘 3.1.b.6 Бизнес-кейс — обучение сотрудников
# ================================================================
before6 = [40, 42, 38, 41, 37, 39]
after6  = [45, 50, 40, 47, 42, 41]

t6, bf10_6, bf01_6 = bayesian_paired_test(before6, after6, alternative="greater")
print("3.1.b.6 training | BF10 =", bf10_6)


# ================================================================
# 📘 3.1.b.7 Бизнес-кейс — ускорение IT системы
# ================================================================
response_before = [320, 300, 310, 330, 305]
response_after  = [250, 270, 260, 240, 265]

t7, bf10_7, bf01_7 = bayesian_paired_test(response_before, response_after, alternative="less")
print("3.1.b.7 faster system | BF10 =", bf10_7)


# ================================================================
# 📘 ИТОГ:
# ------------------------------------------------
# ✔ Байесовский аналог paired t-test
# ✔ Показывает силу доказательств (BF10, BF01)
# ✔ Работает при малых выборках
# ✔ Не требует нормальности разностей
# ✔ Полезен в бизнесе: «насколько вероятно улучшение?»
# ✔ Поддерживает one-sided и two-sided
# ------------------------------------------------
# BF10 > 3  — умеренные доказательства за H1
# BF10 > 10 — сильные доказательства за H1
# BF10 > 30 — очень сильные
# BF01 > 3  — доказательства за отсутствие эффекта
# ================================================================
```

```python
# ================================================================
# 📘 3.2 Wilcoxon signed-rank test (для зависимых ненормальных данных)
# ================================================================
# Использовать когда:
#   • две зависимые выборки (до/после)
#   • разности НЕ нормальны
#   • есть выбросы, шумы
#   • важны знак и величина изменений
#
# Проверяет:
#   H0: распределение разностей симметрично вокруг 0
#   H1: распределение сдвинуто > 0, < 0 или ≠ 0
#
# Альтернативы:
#   'two-sided'
#   'greater' → after > before
#   'less'    → after < before
#
# Отличие от sign-test:
#   sign-test учитывает только знак (+/-)
#   Wilcoxon учитывает и знак, и ранги → мощнее
# ================================================================

import numpy as np
from scipy.stats import wilcoxon


def wilcoxon_test(before, after, alternative="two-sided"):
    """
    Обёртка над scipy.stats.wilcoxon.
    Возвращает: statistic, p_value
    """

    before = np.asarray(before)
    after  = np.asarray(after)

    res = wilcoxon(after - before, alternative=alternative)
    return res.statistic, res.pvalue


# ================================================================
# 📘 3.2.1 Пример: число ошибок снизилось (after < before)
# ================================================================
before1 = [30, 32, 34, 31]
after1  = [25, 26, 28, 29]

stat1, p1 = wilcoxon_test(before1, after1, alternative="less")
print("3.2.1 fewer errors after UI update:", p1)


# ================================================================
# 📘 3.2.2 Проверяем рост показателя (after > before)
# ================================================================
before2 = [100, 120, 110, 115]
after2  = [140, 150, 160, 155]

stat2, p2 = wilcoxon_test(before2, after2, alternative="greater")
print("3.2.2 performance improved:", p2)


# ================================================================
# 📘 3.2.3 Двусторонний тест (любой сдвиг)
# ================================================================
before3 = [45, 47, 44, 50, 48]
after3  = [47, 48, 46, 49, 47]

stat3, p3 = wilcoxon_test(before3, after3)
print("3.2.3 two-sided:", p3)


# ================================================================
# 📘 3.2.4 Малые выборки (Wilcoxon — лучший выбор)
# ================================================================
before4 = [10, 12, 13]
after4  = [11, 12, 14]

stat4, p4 = wilcoxon_test(before4, after4)
print("3.2.4 small sample:", p4)


# ================================================================
# 📘 3.2.5 Присутствуют выбросы (Wilcoxon устойчив)
# ================================================================
before5 = [100, 102, 101, 500]
after5  = [95, 100, 90, 480]

stat5, p5 = wilcoxon_test(before5, after5, alternative="less")
print("3.2.5 outliers:", p5)


# ================================================================
# 📘 3.2.6 Бизнес-кейс — обучение сотрудников
# ================================================================
before6 = [40, 42, 38, 41, 37, 39]
after6  = [45, 50, 40, 47, 42, 41]

stat6, p6 = wilcoxon_test(before6, after6, alternative="greater")
print("3.2.6 training effect:", p6)


# ================================================================
# 📘 3.2.7 Бизнес-кейс — ускорение работы IT системы
# ================================================================
response_before = [320, 300, 310, 330, 305]
response_after  = [250, 270, 260, 240, 265]

stat7, p7 = wilcoxon_test(response_before, response_after, alternative="less")
print("3.2.7 faster system:", p7)


# ================================================================
# 📘 3.2.8 Тестирование двух алгоритмов на тех же данных
# ================================================================
algo_A = [0.87, 0.88, 0.86, 0.89]
algo_B = [0.90, 0.92, 0.91, 0.93]

stat8, p8 = wilcoxon_test(algo_A, algo_B, alternative="greater")
print("3.2.8 algo_B better:", p8)


# ================================================================
# 📘 ИТОГ:
# ------------------------------------------------
# ✔ two-sided / greater / less
# ✔ идеален для зависимых ненормальных данных
# ✔ устойчив к выбросам
# ✔ работает при малых выборках
# ✔ мощнее sign-test
# ✔ лучший выбор до/после, когда t-test неприменим
# ------------------------------------------------
# Wilcoxon — основной непараметрический тест для зависимых данных.
# ================================================================
```

```python
# ================================================================
# 📘 3.2.a Sign-test (paired)
# ================================================================
# Использовать когда:
#   • две зависимые группы (до/после)
#   • разности НЕ симметричны
#   • сильные выбросы, шумы
#   • величина изменений НЕ важна
#   • Wilcoxon неприменим (требует симметричных разностей)
#
# Применение:
#   • дефекты до/после
#   • изменение отказов, ошибок, аварий
#   • измерения с редкими большими скачками
#
# Проверяет:
#   H0: median(after - before) = 0
#   H1: >0 (greater), <0 (less), ≠0 (two-sided)
#
# Примечание:
#   Sign-test использует ТОЛЬКО знаки (+ / -), игнорирует величину.
# ================================================================

import numpy as np
from statsmodels.stats.descriptivestats import sign_test


def paired_sign_test(before, after, alternative="two-sided"):
    """
    Парный sign-test.
    Возвращает:
        test_statistic, p_value
    """

    before = np.asarray(before)
    after  = np.asarray(after)

    diff = after - before

    # mu0=0 → проверяем median(diff)=0
    stat, p = sign_test(diff, mu0=0)

    # statsmodels всегда даёт two-sided p,
    # поэтому вручную делаем направленные варианты
    if alternative == "two-sided":
        return stat, p

    # Доля положительных изменений
    pos = np.sum(diff > 0)
    neg = np.sum(diff < 0)

    if alternative == "greater":   # after > before
        p_one = p / 2 if pos > neg else 1 - p/2
        return stat, p_one

    if alternative == "less":      # after < before
        p_one = p / 2 if neg > pos else 1 - p/2
        return stat, p_one

    return stat, p


# ================================================================
# 📘 3.2.a.1 Пример: дефекты уменьшились (after < before)
# ================================================================
before1 = [10, 12, 8, 9, 11]
after1  = [8, 10, 7, 9, 8]

s1, p1 = paired_sign_test(before1, after1, alternative="less")
print("3.2.a.1 defects decreased:", p1)


# ================================================================
# 📘 3.2.a.2 Пример: производительность выросла (after > before)
# ================================================================
before2 = [40, 42, 38, 41, 37]
after2  = [45, 43, 39, 47, 39]

s2, p2 = paired_sign_test(before2, after2, alternative="greater")
print("3.2.a.2 performance ↑:", p2)


# ================================================================
# 📘 3.2.a.3 Двусторонний тест (любой сдвиг)
# ================================================================
before3 = [12, 15, 11, 14]
after3  = [10, 18, 14, 13]

s3, p3 = paired_sign_test(before3, after3)
print("3.2.a.3 two-sided:", p3)


# ================================================================
# 📘 3.2.a.4 Сильные выбросы → sign-test не чувствителен
# ================================================================
before4 = [10, 12, 13, 1000]
after4  = [11, 11, 12,  999]  # огромные выбросы

s4, p4 = paired_sign_test(before4, after4, alternative="less")
print("3.2.a.4 outliers ok:", p4)


# ================================================================
# 📘 3.2.a.5 Малые выборки (идеально для sign-test)
# ================================================================
before5 = [5, 6, 7]
after5  = [4, 6, 6]

s5, p5 = paired_sign_test(before5, after5, alternative="less")
print("3.2.a.5 small sample:", p5)


# ================================================================
# 📘 3.2.a.6 Бизнес-кейс — аварии до/после
# ================================================================
before6 = [3, 5, 2, 4, 6]
after6  = [1, 3, 2, 2, 5]

s6, p6 = paired_sign_test(before6, after6, alternative="less")
print("3.2.a.6 failures decreased:", p6)


# ================================================================
# 📘 3.2.a.7 Бизнес-кейс — продажи после обучения
# ================================================================
before7 = [100, 120, 110, 95, 130]
after7  = [120, 130, 108, 100, 135]

s7, p7 = paired_sign_test(before7, after7, alternative="greater")
print("3.2.a.7 sales increased:", p7)


# ================================================================
# 📘 ИТОГ:
# ------------------------------------------------
# ✔ работает при асимметричных распределениях
# ✔ полностью устойчив к выбросам
# ✔ лучший выбор при «грязных» данных
# ✔ идеален для маленьких выборок
# ✔ направление: >, <, ≠
# ✔ бизнес-кейсы: дефекты, аварии, простои, продажи
# ------------------------------------------------
# Sign-test — самый устойчивый (но слабый) тест для зависимых данных.
# Если Wilcoxon нельзя → sign-test это лучший вариант.
# ================================================================
```

```python
# ================================================================
# 📘 3.2.b Paired Bootstrap test
# ================================================================
# Использовать когда:
#   • две зависимые группы (до/после)
#   • разности ненормальные
#   • ОЧЕНЬ много выбросов, хаос в данных
#   • Wilcoxon и t-test дают некорректные результаты
#
# Проверяет:
#   H0: mean(after - before) = 0
#   H1: greater / less / two-sided
#
# Плюсы bootstrap:
#   ✔ не требует нормальности
#   ✔ устойчив к выбросам
#   ✔ работает при малых выборках
#   ✔ сравнивает эффекты напрямую
# ================================================================

import numpy as np


def paired_bootstrap_test(before, after, n_boot=20000, alternative="two-sided"):
    """
    Bootstrap-тест для двух зависимых выборок.
    
    Возвращает:
        observed_diff, p_value
    """
    before = np.asarray(before)
    after  = np.asarray(after)

    diff = after - before
    n = len(diff)

    observed = np.mean(diff)

    # bootstrap: выборка с возвращением
    boot_means = np.zeros(n_boot)
    for i in range(n_boot):
        sample = np.random.choice(diff, size=n, replace=True)
        boot_means[i] = np.mean(sample)

    # вычисляем p-value
    if alternative == "two-sided":
        p = np.mean(np.abs(boot_means) >= np.abs(observed))

    elif alternative == "greater":
        p = np.mean(boot_means >= observed)

    elif alternative == "less":
        p = np.mean(boot_means <= observed)

    return observed, p


# ================================================================
# 📘 3.2.b.1 Пример: время обработки звонков снизилось
# ================================================================
before1 = [300, 280, 350, 500, 320]  # нестабильные, много шумов
after1  = [260, 250, 310, 420, 300]

obs1, p1 = paired_bootstrap_test(before1, after1, alternative="less")
print("3.2.b.1 calls faster after training:", p1, "| diff =", obs1)


# ================================================================
# 📘 3.2.b.2 after > before (увеличение показателя)
# ================================================================
before2 = [100, 120, 110, 115]
after2  = [140, 150, 160, 155]

obs2, p2 = paired_bootstrap_test(before2, after2, alternative="greater")
print("3.2.b.2 after > before:", p2, "| diff =", obs2)


# ================================================================
# 📘 3.2.b.3 Двусторонний тест при нестабильных данных
# ================================================================
before3 = [45, 47, 44, 90, 200]
after3  = [47, 48, 46, 95, 210]

obs3, p3 = paired_bootstrap_test(before3, after3)
print("3.2.b.3 two-sided:", p3, "| diff =", obs3)


# ================================================================
# 📘 3.2.b.4 Малые выборки
# ================================================================
before4 = [10, 12, 13]
after4  = [11, 12, 14]

obs4, p4 = paired_bootstrap_test(before4, after4)
print("3.2.b.4 small sample:", p4, "| diff =", obs4)


# ================================================================
# 📘 3.2.b.5 Огромные выбросы — bootstrap устойчив
# ================================================================
before5 = [100, 102, 101, 3000]
after5  = [90, 95, 92, 2900]

obs5, p5 = paired_bootstrap_test(before5, after5, alternative="less")
print("3.2.b.5 outliers:", p5, "| diff =", obs5)


# ================================================================
# 📘 3.2.b.6 Бизнес-кейс — обучение операторов call-центра
# ================================================================
before6 = [40, 52, 38, 41, 90, 200]
after6  = [35, 48, 36, 39, 80, 150]

obs6, p6 = paired_bootstrap_test(before6, after6, alternative="less")
print("3.2.b.6 operators faster:", p6, "| diff =", obs6)


# ================================================================
# 📘 3.2.b.7 Бизнес-кейс — повышение продаж после тренинга
# ================================================================
before7 = [100, 120, 110, 95, 300]
after7  = [120, 130, 108, 100, 320]

obs7, p7 = paired_bootstrap_test(before7, after7, alternative="greater")
print("3.2.b.7 sales increased:", p7, "| diff =", obs7)


# ================================================================
# 📘 ИТОГ:
# ------------------------------------------------
# ✔ лучший тест для зависимых данных с выбросами
# ✔ two-sided / greater / less
# ✔ не требует нормальности
# ✔ работает при малых выборках
# ✔ учитывает величину изменений (в отличие от sign-test)
# ✔ идеален для метрик со скачками, простоями, аномалиями
# ------------------------------------------------
# Bootstrap paired test — мощный и максимально устойчивый метод.
# ================================================================
```

```python
# ================================================================
# 📘 4.1 ONE-WAY ANOVA (три и более независимых групп)
# ================================================================
# Когда использовать:
#   • ≥ 3 независимых групп
#   • данные нормальные (или примерно нормальные)
#   • дисперсии примерно равны
#
# Проверяет:
#   H0: mean(g1) = mean(g2) = mean(g3)
#   H1: хотя бы одно среднее отличается
#
# Важно:
#   ANOVA *не показывает*, какие группы отличаются → нужен post-hoc (Tukey)
#
# Альтернативы:
#   • Welch ANOVA (если дисперсии НЕ равны)
#   • Kruskal–Wallis (если нет нормальности)
# ================================================================

import numpy as np
from scipy import stats
import pandas as pd


# ================================================================
# 📘 4.1.a ФУНКЦИЯ: One-way ANOVA
# ================================================================
def one_way_anova(*groups):
    """
    Классический однофакторный ANOVA.

    Параметры:
        groups — произвольное количество массивов значений (g1, g2, g3, ...)

    Возвращает:
        F-statistic, p-value
    """
    return stats.f_oneway(*groups)


# ================================================================
# 📘 4.1.b Проверки перед ANOVA
# ================================================================
def check_assumptions(*groups):
    """
    Проверки:
      1) нормальность → Shapiro–Wilk
      2) равенство дисперсий → Levene test
    """
    groups = [np.asarray(g) for g in groups]

    # p-values тестов Шапиро (по одному на группу)
    shapiro_ps = [stats.shapiro(g)[1] for g in groups]

    # p-value теста Левена на равенство дисперсий
    levene_p = stats.levene(*groups)[1]

    return {
        "shapiro_ps": shapiro_ps,
        "levene_p": levene_p
    }


# ================================================================
# 📘 4.1.c Пост-хок анализ (Tukey HSD)
# ================================================================
def tukey_hsd(groups, labels):
    """
    Пост-хок тест: показывает, какие пары групп отличаются.
    """
    from statsmodels.stats.multicomp import pairwise_tukeyhsd

    # объединяем данные и подписи групп
    data = np.concatenate(groups)
    lab  = np.concatenate([[label] * len(g) for g, label in zip(groups, labels)])

    return pairwise_tukeyhsd(data, lab)


# ================================================================
# 📘 4.1.1 Пример: три района города (классический кейс)
# ================================================================
g1 = [20, 22, 21, 23, 19]  # район A
g2 = [25, 27, 26, 30, 24]  # район B
g3 = [31, 33, 35, 32, 30]  # район C

# Проверяем допущения
print("4.1.1 CHECK ASSUMPTIONS:", check_assumptions(g1, g2, g3))

# Проводим ANOVA
F1, p1 = one_way_anova(g1, g2, g3)
print("4.1.1 ANOVA A/B/C:", "p-value =", p1)

# Если p < 0.05 → делаем пост-хок
if p1 < 0.05:
    print(tukey_hsd([g1, g2, g3], ["A", "B", "C"]))


# ================================================================
# 📘 4.1.2 Пример: средние продажи в магазинах
# ================================================================
shop1 = [100, 120, 115, 130, 110]
shop2 = [150, 160, 155, 170, 158]
shop3 = [90, 85, 80, 95, 92]

F2, p2 = one_way_anova(shop1, shop2, shop3)
print("4.1.2 sales ANOVA:", "p-value =", p2)


# ================================================================
# 📘 4.1.3 Пример: время ответа серверов
# ================================================================
dc1 = np.random.normal(100, 5, 50)
dc2 = np.random.normal(120, 7, 50)
dc3 = np.random.normal(140, 8, 50)

F3, p3 = one_way_anova(dc1, dc2, dc3)
print("4.1.3 server latency ANOVA:", "p-value =", p3)


# ================================================================
# 📘 4.1.4 Пример: эффективность трёх видов обучения
# ================================================================
train_A = [80, 82, 81, 85, 83]
train_B = [88, 90, 91, 87, 89]
train_C = [70, 68, 72, 69, 71]

F4, p4 = one_way_anova(train_A, train_B, train_C)
print("4.1.4 training ANOVA:", "p-value =", p4)


# ================================================================
# 📘 4.1.5 Малые выборки
# ================================================================
s1 = [10, 11, 9]
s2 = [15, 16, 14]
s3 = [20, 19, 18]

F5, p5 = one_way_anova(s1, s2, s3)
print("4.1.5 small samples:", "p-value =", p5)


# ================================================================
# 📘 4.1.6 Пример: одинаковые средние
# ================================================================
a1 = [50, 51, 49, 52]
a2 = [48, 50, 52, 49]
a3 = [51, 53, 50, 49]

F6, p6 = one_way_anova(a1, a2, a3)
print("4.1.6 equal means:", "p-value =", p6)


# ================================================================
# 📘 ИТОГ
# ------------------------------------------------
# ✔ ANOVA — сравнение средних 3+ независимых групп
# ✔ Требования:
#       — нормальность
#       — равенство дисперсий
# ✔ Post-hoc Tukey показывает, какие группы отличаются
#
# Альтернативы:
#   4.2 Welch ANOVA (если дисперсии не равны)
#   4.3 Kruskal–Wallis (если нет нормальности)
# ================================================================
```

```python
# ================================================================
# 📘 4.1.a WELCH ANOVA (разные дисперсии, 3+ независимых групп)
# ================================================================
# Когда использовать:
#   • группы независимы
#   • ≥ 3 групп
#   • данные нормальные или почти нормальные
#   • дисперсии НЕ равны (Levene test → p < 0.05)
#
# Проверяет:
#   H0: mean(g1) = mean(g2) = mean(g3)
#   H1: хотя бы одно среднее отличается
#
# Преимущества:
#   ✔ корректен при разных дисперсиях
#   ✔ корректен при разных размерах групп
#   ✔ всегда безопаснее, чем обычный ANOVA
#
# SciPy НЕ содержит Welch ANOVA → используется statsmodels
# ================================================================

import numpy as np
import pandas as pd
import statsmodels.api as sm
from statsmodels.formula.api import ols


# ================================================================
# 📘 ФУНКЦИЯ: Welch ANOVA
# ================================================================
def welch_anova(groups, labels):
    """
    Выполняет Welch's ANOVA для 3+ групп.

    groups  — список массивов данных: [g1, g2, g3, ...]
    labels  — список названий групп:   ["A", "B", "C"]

    Возвращает таблицу ANOVA Welch'а.
    """

    # собираем данные в DataFrame в формате long
    df = pd.DataFrame({
        "value": np.concatenate(groups),
        "group": np.concatenate([[lab] * len(g) for g, lab in zip(groups, labels)])
    })

    # модель OLS
    model = ols("value ~ group", data=df).fit()

    # Welch ANOVA
    welch_table = sm.stats.anova_lm(model, typ=2, robust='hc3')

    return welch_table


# ================================================================
# 📘 4.1.a.1 Пример — зарплаты в 3 компаниях (разные дисперсии)
# ================================================================
# Компания A — зарплаты стабильные
# Компания B — большие колебания
# Компания C — очень разнородный персонал

A = np.random.normal(2000,  50, 200)   # низкая дисперсия
B = np.random.normal(2200, 200, 200)   # высокая дисперсия
C = np.random.normal(2500, 400, 200)   # очень высокая дисперсия

result1 = welch_anova([A, B, C], ["A", "B", "C"])
print("\n4.1.a.1 Salaries (Welch ANOVA):\n", result1)


# ================================================================
# 📘 4.1.a.2 Пример — скорость отклика серверов в 3 дата-центрах
# ================================================================
dc1 = np.random.normal(100, 5, 50)
dc2 = np.random.normal(120, 15, 50)
dc3 = np.random.normal(140, 30, 50)

result2 = welch_anova([dc1, dc2, dc3], ["DC1", "DC2", "DC3"])
print("\n4.1.a.2 Server latency (Welch ANOVA):\n", result2)


# ================================================================
# 📘 4.1.a.3 Пример — производительность отделов
# ================================================================
depA = np.random.normal(80, 10, 100)
depB = np.random.normal(90, 30, 100)
depC = np.random.normal(95, 50, 100)

result3 = welch_anova([depA, depB, depC], ["A", "B", "C"])
print("\n4.1.a.3 Productivity (Welch ANOVA):\n", result3)


# ================================================================
# 📘 ИТОГ:
# ------------------------------------------------
# ✔ Welch ANOVA нужна, когда дисперсии НЕ равны
# ✔ Полностью заменяет обычную ANOVA при гетероскедастичности
# ✔ Используется в исследовании зарплат, метрик качества, медицинских данных
# ✔ Работает для любых 3+ групп с нормальными распределениями
# ------------------------------------------------
# Следующий шаг (если p < 0.05):
#   → выполняем Games–Howell post-hoc test
#     (идеален при неравных дисперсиях и размерах)
# ================================================================
```

```python
# ================================================================
# 📘 4.1.b PERMUTATION ANOVA (перестановочный ANOVA)
# ================================================================
# Когда использовать:
#   • 3 и более независимых групп
#   • распределения НЕ нормальные
#   • дисперсии НЕ равны
#   • выбросы, тяжёлые хвосты
#   • когда хотим максимально честный non-parametric ANOVA
#
# Проверяет:
#   H0: средние значений во всех группах одинаковы
#   H1: хотя бы одна группа отличается
#
# Преимущества:
#   ✔ не требует нормальности
#   ✔ работает при разных дисперсиях
#   ✔ работает при разном размере групп
#   ✔ корректно обрабатывает выбросы
#
# Недостаток:
#   — требует вычислительных ресурсов (особенно если perm=100k)
# ================================================================

import numpy as np


# ================================================================
# ФУНКЦИЯ: вычисление F-статистики (как в ANOVA)
# ================================================================
def calc_f_stat(groups):
    """
    groups — список массивов (g1, g2, g3, ...)

    Возвращает F-статистику:
        межгрупповая вариация / внутригрупповая вариация
    """

    all_data = np.concatenate(groups)
    grand_mean = np.mean(all_data)

    # межгрупповая сумма квадратов
    ss_between = sum(len(g) * (np.mean(g) - grand_mean)**2 for g in groups)

    # внутригрупповая сумма квадратов
    ss_within = sum(sum((g - np.mean(g))**2) for g in groups)

    df_between = len(groups) - 1
    df_within = len(all_data) - len(groups)

    ms_between = ss_between / df_between
    ms_within  = ss_within  / df_within

    return ms_between / ms_within


# ================================================================
# 📘 ФУНКЦИЯ: Permutation ANOVA
# ================================================================
def permutation_anova(groups, n_perm=5000, random_state=42):
    """
    groups : list of arrays, например:
        [g1, g2, g3]

    n_perm : число перестановок (5000 обычно достаточно)

    Возвращает:
        F_real  — реальный F
        p_value — доля перестановок, где F_perm ≥ F_real
    """

    rng = np.random.default_rng(random_state)
    groups = [np.asarray(g) for g in groups]

    # реальный F-статистик
    F_real = calc_f_stat(groups)

    # объединяем все данные
    pooled = np.concatenate(groups)

    counts = [len(g) for g in groups]
    n_total = len(pooled)

    # считаем перестановки
    F_perm_values = []
    for _ in range(n_perm):
        shuffled = rng.permutation(pooled)

        # распиливаем обратно по размерам групп
        perm_groups = []
        idx = 0
        for n in counts:
            perm_groups.append(shuffled[idx:idx+n])
            idx += n

        F_perm = calc_f_stat(perm_groups)
        F_perm_values.append(F_perm)

    F_perm_values = np.array(F_perm_values)

    # p-value = доля перестановок, где F_perm ≥ F_real
    p_value = np.mean(F_perm_values >= F_real)

    return F_real, p_value


# ================================================================
# 📘 4.1.b.1 Пример — доход 3 подразделений (нет нормальности)
# ================================================================
depA = np.random.gamma(2, 2000, 300)   # сильно асимметрично
depB = np.random.gamma(5, 1500, 300)
depC = np.random.gamma(3, 2500, 300)

F1, p1 = permutation_anova([depA, depB, depC], n_perm=7000)
print("\n4.1.b.1 income departments → p =", p1, "| F_real =", F1)


# ================================================================
# 📘 4.1.b.2 Пример — время ответа серверов (разные дисперсии)
# ================================================================
dc1 = np.random.exponential(1.0, 200)
dc2 = np.random.exponential(2.0, 200)
dc3 = np.random.exponential(3.0, 200)

F2, p2 = permutation_anova([dc1, dc2, dc3])
print("4.1.b.2 server latency → p =", p2)


# ================================================================
# 📘 4.1.b.3 Пример — качество продукта в 3 филиалах
# ================================================================
q1 = np.random.lognormal(mean=1.5, sigma=0.4, size=150)
q2 = np.random.lognormal(mean=1.4, sigma=0.7, size=150)
q3 = np.random.lognormal(mean=1.8, sigma=0.6, size=150)

F3, p3 = permutation_anova([q1, q2, q3])
print("4.1.b.3 product quality → p =", p3)


# ================================================================
# 📘 Итог:
# ------------------------------------------------
# ✔ Permutation ANOVA — лучший выбор, если:
#      — нет нормальности
#      — разные дисперсии
#      — есть выбросы
#      — хотите честный non-parametric ANOVA
#
# ✔ Альтернативы:
#      4.1 One-way ANOVA       (нормальные данные)
#      4.1.a Welch ANOVA       (разные дисперсии)
#      4.1.c Kruskal–Wallis    (сравнивает распределения, не средние)
# ------------------------------------------------
# ================================================================
```

```python
# ================================================================
# 📘 4.1.c BROWN–FORSYTHE ANOVA
# ================================================================
# Когда использовать:
#   • минимальные предположения о данных
#   • сильные выбросы
#   • группы имеют разные дисперсии
#   • ANOVA и Welch ANOVA ненадёжны
#
# Проверяет:
#   H0: медианные отклонения одинаковы во всех группах
#   H1: хотя бы одна группа отличается
#
# Метод:
#   1) Для каждого значения считаем |x - median(group)|
#   2) На этих величинах выполняем обычную ANOVA
#
# Преимущества:
#   ✔ устойчив к выбросам
#   ✔ не требует равенства дисперсий
#   ✔ корректен при отсутствии нормальности
# ================================================================

import numpy as np
from scipy.stats import f


# ================================================================
# 📘 Функция: Brown–Forsythe ANOVA
# ================================================================
def brown_forsythe_anova(groups):
    """
    Выполняет Brown–Forsythe ANOVA для 3+ независимых групп.

    groups — список массивов данных [g1, g2, g3, ...]

    Возвращает:
        F, p_value
    """

    groups = [np.asarray(g) for g in groups]

    # 1. Абсолютные отклонения от медианы своей группы
    deviations = [np.abs(g - np.median(g)) for g in groups]

    # 2. Классическая ANOVA НА ОТКЛОНЕНИЯХ
    all_dev = np.concatenate(deviations)
    grand_mean = np.mean(all_dev)

    # Межгрупповая сумма квадратов
    ss_between = sum(len(d) * (np.mean(d) - grand_mean)**2 for d in deviations)

    # Внутригрупповая сумма квадратов
    ss_within = sum(sum((d - np.mean(d))**2) for d in deviations)

    df_between = len(groups) - 1
    df_within = len(all_dev) - len(groups)

    ms_between = ss_between / df_between
    ms_within  = ss_within  / df_within

    F = ms_between / ms_within

    # p-value через F-распределение
    p_value = 1 - f.cdf(F, df_between, df_within)

    return F, p_value


# ================================================================
# 📘 4.1.c.1 Пример — выработка 3 производственных линий
#     (нестабильность + выбросы)
# ================================================================

lineA = np.concatenate([np.random.normal(100, 5, 100), [200]])   # выброс
lineB = np.random.normal(120, 20, 120)                          # высокая дисперсия
lineC = np.random.normal(90, 10, 150)

F1, p1 = brown_forsythe_anova([lineA, lineB, lineC])
print("\n4.1.c.1 Production lines → p =", p1, "| F =", F1)


# ================================================================
# 📘 4.1.c.2 Пример — время отклика 3 серверов (нестабильность)
# ================================================================

dc1 = np.concatenate([np.random.normal(100, 5, 200), [400]])
dc2 = np.random.normal(150, 30, 200)
dc3 = np.random.normal(120, 15, 200)

F2, p2 = brown_forsythe_anova([dc1, dc2, dc3])
print("4.1.c.2 Server latency → p =", p2)


# ================================================================
# 📘 4.1.c.3 Пример — качество партий продукции
# ================================================================

batchA = np.random.lognormal(mean=3.2, sigma=0.2, size=300)
batchB = np.random.lognormal(mean=3.25, sigma=0.4, size=300)
batchC = np.random.lognormal(mean=3.0, sigma=0.5, size=300)

F3, p3 = brown_forsythe_anova([batchA, batchB, batchC])
print("4.1.c.3 Batch quality → p =", p3)


# ================================================================
# 📘 4.1.c.4 Малые выборки + выбросы
# ================================================================

small1 = [10, 11, 9, 200]
small2 = [12, 13, 14, 15]
small3 = [8, 7, 6, 9]

F4, p4 = brown_forsythe_anova([small1, small2, small3])
print("4.1.c.4 Small n + outliers → p =", p4)


# ================================================================
# 📘 Итог:
# ------------------------------------------------
# ✔ Brown–Forsythe — лучший тест при:
#       → выбросах
#       → гетерогенных дисперсиях
#       → отсутствии нормальности
# ✔ Альтернативы:
#       4.1.a Welch ANOVA        (тоже для разных дисперсий)
#       4.1.b Permutation ANOVA  (полностью non-parametric)
#       4.1.d Kruskal–Wallis     (сравнивает распределения)
# ------------------------------------------------
# ================================================================
```

```python
# ================================================================
# 📘 4.2 Kruskal–Wallis H-test (3+ независимых групп, ненормальные)
# ================================================================
# Когда использовать:
#   • 3 и более независимых групп
#   • данные НЕ нормальные
#   • выбросы
#   • разные дисперсии
#   • распределения асимметричные
#
# Проверяет:
#   H0: распределения всех групп одинаковы
#   H1: хотя бы одна группа отличается
#
# Особенности:
#   ✔ непараметрический аналог ANOVA
#   ✔ работает на рангах, а не на числах
#   ✔ устойчив к выбросам
#   ✔ не требует нормальности
#   ✔ не требует равенства дисперсий
#
# Пример применения:
#   Сравнить оценки клиентов в трёх филиалах сети.
#
# Альтернативы:
#   • Welch ANOVA — если данные нормальные, но дисперсии разные
#   • Brown–Forsythe ANOVA — если выбросы + разные дисперсии
#   • Permutation ANOVA — наиболее честный непараметрический тест
# ================================================================

import numpy as np
from scipy.stats import kruskal


# ================================================================
# 📘 ФУНКЦИЯ: Kruskal–Wallis H-test
# ================================================================
def kruskal_test(*groups):
    """
    Выполняет Kruskal–Wallis тест для k независимых групп.

    Параметры:
        groups — произвольное число массивов данных

    Возвращает:
        H-statistic, p-value
    """
    return kruskal(*groups)


# ================================================================
# 📘 4.2.1 Оценки клиентов в 3 филиалах (типовой кейс)
# ================================================================
branch_A = [8, 9, 9, 10, 9, 8]
branch_B = [6, 7, 6, 7, 6, 7]
branch_C = [5, 6, 4, 7, 5, 6]

H1, p1 = kruskal_test(branch_A, branch_B, branch_C)
print("4.2.1 Client scores → p =", p1, "| H =", H1)


# ================================================================
# 📘 4.2.2 Устойчивость к выбросам
# ================================================================
g1 = [10, 12, 11, 13, 200]    # выброс
g2 = [8, 7, 9, 10, 11]
g3 = [15, 14, 16, 15, 17]

H2, p2 = kruskal_test(g1, g2, g3)
print("4.2.2 Outliers → p =", p2)


# ================================================================
# 📘 4.2.3 Разные дисперсии
# ================================================================
high_var   = np.random.normal(50, 20, 200)
medium_var = np.random.normal(55,  5, 200)
low_var    = np.random.normal(60,  1, 200)

H3, p3 = kruskal_test(high_var, medium_var, low_var)
print("4.2.3 Unequal variances → p =", p3)


# ================================================================
# 📘 4.2.4 Ненормальные, асимметричные распределения
# ================================================================
exp1 = np.random.exponential(scale=1, size=300)
exp2 = np.random.exponential(scale=2, size=300)
exp3 = np.random.exponential(scale=3, size=300)

H4, p4 = kruskal_test(exp1, exp2, exp3)
print("4.2.4 Exponential → p =", p4)


# ================================================================
# 📘 4.2.5 Малые выборки
# ================================================================
small1 = [1, 2, 3]
small2 = [4, 5, 6]
small3 = [2, 2, 3]

H5, p5 = kruskal_test(small1, small2, small3)
print("4.2.5 Small samples → p =", p5)


# ================================================================
# 📘 4.2.6 Бизнес-кейс — скорость обработки заявок
# ================================================================
dept_A = np.random.gamma(2, 10, 300)
dept_B = np.random.gamma(3,  9, 300)
dept_C = np.random.gamma(2, 12, 300)

H6, p6 = kruskal_test(dept_A, dept_B, dept_C)
print("4.2.6 Processing speed → p =", p6)


# ================================================================
# 📘 4.2.7 Бизнес-кейс — сравнение прибыли магазинов
# ================================================================
profit_A = np.random.lognormal(mean=3,   sigma=0.3, size=400)
profit_B = np.random.lognormal(mean=3.2, sigma=0.3, size=400)
profit_C = np.random.lognormal(mean=3.1, sigma=0.5, size=400)

H7, p7 = kruskal_test(profit_A, profit_B, profit_C)
print("4.2.7 Shop profit → p =", p7)


# ================================================================
# 📘 Итог 4.2 Kruskal–Wallis
# ------------------------------------------------
# ✔ лучший тест для 3+ групп, когда данные НЕ нормальные
# ✔ устойчив к выбросам
# ✔ допускает неравные дисперсии
# ✔ не требует одинакового размера выборок
# ✔ работает на рангах → меньше чувствителен к шуму
#
# Альтернативы:
#   • 4.1.c Brown–Forsythe — устойчив к выбросам, сравнивает отклонения
#   • 4.1.b Permutation ANOVA — полностью непараметрический
#   • 4.1.a Welch ANOVA — если нормальность умеренная
# ------------------------------------------------
# ================================================================
```

```python
# ================================================================
# 📘 4.2.a Median test (k-sample) — сравнение медиан 3+ групп
# ================================================================
# Когда использовать:
#   • хотим сравнить ИМЕННО медианы (а не распределения)
#   • 3 и более независимых групп
#   • данные НЕ нормальны
#   • сильные выбросы
#   • распределения сильно «грязные» или имеют разные хвосты
#
# Проверяет:
#   H0: медианы всех групп равны
#   H1: хотя бы одна медиана отличается
#
# Метод:
#   1) находим общую медиану для всех данных
#   2) строим таблицу “выше/ниже медианы” для каждой группы
#   3) применяем χ²-тест для таблицы
#
# Преимущества:
#   ✔ только медиана — не чувствителен к выбросам
#   ✔ не требует нормальности
#   ✔ устойчив к разным дисперсиям
#
# Недостатки:
#   — теряет много информации (использует бинаризацию)
#   — менее мощный, чем Kruskal–Wallis
#
# Пример применения:
#   сравнить медианные продажи в трёх магазинах
# ================================================================

import numpy as np
from scipy.stats import median_test


# ================================================================
# 📘 ФУНКЦИЯ: k-sample median test
# ================================================================
def median_test_k_sample(*groups):
    """
    Обёртка над scipy.stats.median_test для k групп.

    Возвращает:
        chi2, p_value, contingency_table, grand_median
    """
    return median_test(*groups)


# ================================================================
# 📘 4.2.a.1 Пример — медианные продажи в трёх магазинах
# ================================================================
shop_A = [100, 120, 110, 130, 115]
shop_B = [80,  90,  85,  95,  100]
shop_C = [150, 160, 170, 155, 165]

chi2_1, p1, tbl1, gm1 = median_test_k_sample(shop_A, shop_B, shop_C)
print("4.2.a.1 Shop sales medians → p =", p1, "| grand median =", gm1)
print("Contingency table:\n", tbl1)


# ================================================================
# 📘 4.2.a.2 Асимметричные распределения, тяжёлые хвосты
# ================================================================
g1 = np.random.exponential(scale=10, size=300)
g2 = np.random.exponential(scale=20, size=300)
g3 = np.random.exponential(scale=30, size=300)

chi2_2, p2, tbl2, gm2 = median_test_k_sample(g1, g2, g3)
print("\n4.2.a.2 Exponential → p =", p2, "| grand median =", gm2)


# ================================================================
# 📘 4.2.a.3 Устойчив к выбросам
# ================================================================
gA = [10, 12, 14, 15, 11, 500]   # выброс
gB = [20, 22, 21, 23, 19]
gC = [30, 31, 29, 33, 32]

chi2_3, p3, tbl3, gm3 = median_test_k_sample(gA, gB, gC)
print("\n4.2.a.3 Outliers → p =", p3, "| grand median =", gm3)


# ================================================================
# 📘 4.2.a.4 Разные дисперсии, разные масштабы
# ================================================================
h1 = np.random.normal(50, 1, 100)
h2 = np.random.normal(70, 20, 100)
h3 = np.random.normal(90, 50, 100)

chi2_4, p4, tbl4, gm4 = median_test_k_sample(h1, h2, h3)
print("\n4.2.a.4 Unequal variances → p =", p4, "| grand median =", gm4)


# ================================================================
# 📘 4.2.a.5 Малые выборки
# ================================================================
s1 = [1, 2, 3]
s2 = [10, 12, 14]
s3 = [8, 9, 7]

chi2_5, p5, tbl5, gm5 = median_test_k_sample(s1, s2, s3)
print("\n4.2.a.5 Small n → p =", p5, "| grand median =", gm5)


# ================================================================
# 📘 4.2.a.6 Бизнес-кейс — медианное время обработки заявок
# ================================================================
dept_A = np.random.gamma(2, 10, 400)
dept_B = np.random.gamma(3,  8, 400)
dept_C = np.random.gamma(2, 12, 400)

chi2_6, p6, tbl6, gm6 = median_test_k_sample(dept_A, dept_B, dept_C)
print("\n4.2.a.6 Processing time → p =", p6, "| median =", gm6)


# ================================================================
# 📘 Итог 4.2.a Median test (k-sample)
# ------------------------------------------------
# ✔ сравнение МЕДИАН 3+ групп (в отличие от Kruskal, который сравнивает распределения)
# ✔ устойчив к выбросам
# ✔ не требует нормальности
# ✔ применим при разных дисперсиях
#
# Альтернативы:
#   • 4.2 Kruskal–Wallis — мощнее, но сравнивает распределения
#   • 4.1.b Permutation ANOVA — мощный непараметрический тест разницы средних
#   • 4.1.c Brown–Forsythe — устойчив к выбросам, работает со средними отклонениями
# ------------------------------------------------
# ================================================================
```

```python
# ================================================================
# 📘 4.2.b Jonckheere–Terpstra test (k independent ordered groups)
# ================================================================
# Данная версия НЕ использует scipy.stats (функция удалена в SciPy 1.11+)
# Полная ручная реализация — работает в Python 3.12–3.13.
#
# Когда использовать:
#   • 3+ независимых групп
#   • группы упорядочены (например: низкий → средний → высокий)
#   • нужно проверить монотонный тренд (рост или падение)
#   • данные не нормальные, есть выбросы
#
# Проверяет:
#   H0: нет монотонного тренда между группами
#   H1: есть тренд (increasing / decreasing)
#
# ================================================================

import numpy as np
from math import sqrt
from scipy.stats import norm   # только для CDF (нормальное распределение)


# ================================================================
# 📘 ФУНКЦИЯ: Jonckheere–Terpstra (ручная реализация)
# ================================================================
def jonckheere_terpstra_test(groups, alternative="increasing"):
    """
    groups — список массивов данных в ПОРЯДКЕ РАНГОВ:
             [group1_low, group2_mid, group3_high]
    
    alternative:
        "increasing" — проверка на рост
        "decreasing" — проверка на падение
        "two-sided"  — любой тренд
    
    Возвращает:
        JT-statistic, p-value
    """

    # Преобразуем в numpy
    groups = [np.asarray(g) for g in groups]
    k = len(groups)

    # --- 1. Вычисляем J — сумму всех S_ij ---
    J = 0
    for i in range(k):
        for j in range(i+1, k):
            # сравниваем каждое значение из группы i с каждым из j
            a = groups[i]
            b = groups[j]
            for x in a:
                for y in b:
                    if x < y:
                        J += 1
                    elif x == y:
                        J += 0.5

    # --- 2. Вычисляем математическое ожидание и дисперсию ---
    n = [len(g) for g in groups]
    N = sum(n)

    # Математическое ожидание
    mean_J = 0.5 * (N**2 - sum(ni**2 for ni in n)) / 2

    # Дисперсия (официальная формула JT)
    var_J = (
        (N * (2*N + 3)) -
        sum(ni * (2*ni + 3) for ni in n)
    ) / 72

    # --- 3. z-статистика ---
    z = (J - mean_J) / sqrt(var_J)

    # Направление альтернативы
    if alternative == "increasing":
        p_value = 1 - norm.cdf(z)
    elif alternative == "decreasing":
        p_value = norm.cdf(z)
    else:  # two-sided
        p_value = 2 * (1 - norm.cdf(abs(z)))

    return J, p_value


# ================================================================
# 📘 4.2.b.1 Удовлетворённость ↑ между филиалами 1 → 2 → 3
# ================================================================
filial_1 = [5, 6, 7, 6, 5]
filial_2 = [6, 7, 8, 7, 6]
filial_3 = [8, 9, 9, 8, 7]

J1, p1 = jonckheere_terpstra_test([filial_1, filial_2, filial_3], alternative="increasing")
print("4.2.b.1 Satisfaction trend (↑) → p =", p1, "| JT =", J1)


# ================================================================
# 📘 4.2.b.2 Тренд убывания
# ================================================================
q1 = [90, 85, 88, 92]
q2 = [70, 75, 72, 68]
q3 = [50, 55, 45, 52]

J2, p2 = jonckheere_terpstra_test([q1, q2, q3], alternative="decreasing")
print("4.2.b.2 Quality trend (↓) → p =", p2, "| JT =", J2)


# ================================================================
# 📘 4.2.b.3 Ненормальные данные: экспоненциальный рост
# ================================================================
g1 = np.random.exponential(scale=2,  size=200)
g2 = np.random.exponential(scale=4,  size=200)
g3 = np.random.exponential(scale=10, size=200)

J3, p3 = jonckheere_terpstra_test([g1, g2, g3], alternative="increasing")
print("4.2.b.3 Exponential trend → p =", p3)


# ================================================================
# 📘 4.2.b.4 Устойчивость к выбросам
# ================================================================
a = [10, 12, 15, 300]
b = [20, 22, 25, 23]
c = [30, 35, 33, 31]

J4, p4 = jonckheere_terpstra_test([a, b, c], alternative="increasing")
print("4.2.b.4 Outliers trend → p =", p4)


# ================================================================
# 📘 Итог 4.2.b:
# ------------------------------------------------
# ✔ лучший тест для УПОРЯДОЧЕННЫХ независимых групп
# ✔ мощнее Kruskal–Wallis при тренде
# ✔ устойчив к выбросам
# ✔ не требует нормальности или равенства дисперсий
#
# Альтернативы:
#   • 4.2 Kruskal–Wallis — порядок групп не важен
#   • 4.2.a Median test — сравнение только медиан
#   • Page L-test — для зависимых групп (повторные измерения)
# ------------------------------------------------
# ================================================================
```

```python
# ================================================================
# 📘 4.2.c Permutation k-sample ANOVA (full nonparametric)
# ================================================================
# Когда использовать:
#   • 3+ независимых групп
#   • нет нормальности
#   • есть выбросы
#   • разные дисперсии
#   • группы могут быть разного размера
#   • хотим максимально честный тест, БЕЗ предположений
#
# Проверяет:
#   H0: средние всех k групп одинаковы
#   H1: хотя бы одно среднее отличается
#
# Принцип:
#   1) считаем реальный F (ANOVA F-statistic)
#   2) многократно перемешиваем данные между группами
#   3) считаем F_perm для каждой перестановки
#   4) p = доля F_perm >= F_real
#
# Преимущества:
#   ✔ не требует никакой параметрической модели
#   ✔ устойчив к выбросам
#   ✔ устойчив к гетероскедастичности
#   ✔ корректен при любом распределении
# ================================================================

import numpy as np


# ================================================================
# 📘 Функция: вычисление F-statistic
# ================================================================
def calc_f_stat(groups):
    """
    Вычисляет F как в ANOVA:
        межгрупповая вариация / внутригрупповая вариация
    """

    groups = [np.asarray(g) for g in groups]
    all_data = np.concatenate(groups)

    grand_mean = np.mean(all_data)

    ss_between = sum(len(g) * (np.mean(g) - grand_mean) ** 2 for g in groups)
    ss_within  = sum(sum((x - np.mean(g)) ** 2 for x in g) for g in groups)

    df_between = len(groups) - 1
    df_within  = len(all_data) - len(groups)

    ms_between = ss_between / df_between
    ms_within  = ss_within  / df_within

    return ms_between / ms_within


# ================================================================
# 📘 ФУНКЦИЯ: Permutation k-sample test
# ================================================================
def permutation_k_sample(groups, n_perm=5000, random_state=42):
    """
    groups  — список независимых массивов, например:
              [g1, g2, g3]

    n_perm — число перестановок (по умолчанию 5000)

    Возвращает:
        F_real, p_value
    """

    rng = np.random.default_rng(random_state)
    groups = [np.asarray(g) for g in groups]

    # реальный F-статистик
    F_real = calc_f_stat(groups)

    # объединяем данные
    pooled = np.concatenate(groups)
    counts = [len(g) for g in groups]

    # массив для хранения перестановочных F
    F_perm_values = np.empty(n_perm)

    for i in range(n_perm):
        shuffled = rng.permutation(pooled)

        # обратно по размерам групп
        perm_groups = []
        idx = 0
        for n in counts:
            perm_groups.append(shuffled[idx:idx + n])
            idx += n

        F_perm_values[i] = calc_f_stat(perm_groups)

    # p-value = доля перестановочных F >= F_real
    p_value = np.mean(F_perm_values >= F_real)

    return F_real, p_value


# ================================================================
# 📘 4.2.c.1 Пример — средняя прибыль трёх складов (без предположений)
# ================================================================
warehouse_A = np.random.lognormal(mean=10, sigma=0.5, size=300)
warehouse_B = np.random.lognormal(mean=10.2, sigma=0.8, size=300)
warehouse_C = np.random.lognormal(mean=9.8, sigma=1.0, size=300)

F1, p1 = permutation_k_sample([warehouse_A, warehouse_B, warehouse_C])
print("4.2.c.1 Warehouses profit → p =", p1, "| F =", F1)


# ================================================================
# 📘 4.2.c.2 Выбросы — perm-test устойчив
# ================================================================
g1 = np.concatenate([np.random.normal(100, 10, 100), [5000]])
g2 = np.random.normal(120, 20, 100)
g3 = np.random.normal(90,  15, 100)

F2, p2 = permutation_k_sample([g1, g2, g3])
print("4.2.c.2 Outliers → p =", p2)


# ================================================================
# 📘 4.2.c.3 Ненормальные распределения
# ================================================================
exp1 = np.random.exponential(1, 200)
exp2 = np.random.exponential(2, 200)
exp3 = np.random.exponential(3, 200)

F3, p3 = permutation_k_sample([exp1, exp2, exp3])
print("4.2.c.3 Exponential → p =", p3)


# ================================================================
# 📘 4.2.c.4 Разные дисперсии
# ================================================================
h1 = np.random.normal(50, 2, 200)
h2 = np.random.normal(50, 20, 200)
h3 = np.random.normal(50, 50, 200)

F4, p4 = permutation_k_sample([h1, h2, h3])
print("4.2.c.4 Unequal variances → p =", p4)


# ================================================================
# 📘 4.2.c.5 Малые выборки
# ================================================================
s1 = [1, 2, 3]
s2 = [5, 6, 7]
s3 = [9, 8, 10]

F5, p5 = permutation_k_sample([s1, s2, s3], n_perm=2000)
print("4.2.c.5 Small samples → p =", p5)


# ================================================================
# 📘 Итог 4.2.c Permutation k-sample
# ------------------------------------------------
# ✔ самый гибкий тест для 3+ независимых групп
# ✔ можно использовать ВСЕГДА (вообще без предположений)
# ✔ устойчив к выбросам, шуму, разным дисперсиям
# ✔ работает при любых распределениях
# ✔ подходит для бизнес-метрик, прибыльности, SLA, времени выполнения
#
# Альтернативы:
#   • 4.2 Kruskal–Wallis — сравнение распределений по рангам
#   • 4.2.a Median test — сравнение медиан
#   • 4.2.b Jonckheere–Terpstra — тест тренда (ordered groups)
# ------------------------------------------------
# ================================================================
```

```python
# ================================================================
# 📘 5.1 Repeated Measures ANOVA (RM-ANOVA)
# ================================================================
# Когда использовать:
#   • 3+ зависимых измерений у тех же субъектов
#   • нормальные данные
#   • нет сильных выбросов
#   • равная дисперсия разностей (сферичность — доп. предпосылка)
#
# Проверяет:
#   H0: средние всех условий равны
#       mean(cond1) = mean(cond2) = mean(cond3) ...
#   H1: хотя бы одно среднее отличается
#
# Используется для:
#   ✔ утро / день / вечер
#   ✔ до / после / через неделю
#   ✔ интерфейс A / B / C на тех же сотрудниках
#
# Альтернативы (если нормальность нарушена):
#   • 5.2 Friedman test (ненормальные)
#   • 5.3 Repeated-measures permutation test (без предположений)
#
# Требуемый формат данных для statsmodels:
#     id | condition | value
# ================================================================

import numpy as np
import pandas as pd
from statsmodels.stats.anova import AnovaRM


# ================================================================
# 📘 Утилита: подготовка DataFrame
# ================================================================
def make_rm_dataframe(values_dict):
    """
    Преобразует словарь формата:
        {
            "cond1": [...],
            "cond2": [...],
            "cond3": [...]
        }
    в DataFrame формата:
        id | condition | value
    """

    df_list = []
    n = len(next(iter(values_dict.values())))

    for cond, vals in values_dict.items():
        df_list.append(pd.DataFrame({
            "id": np.arange(n),
            "condition": cond,
            "value": vals
        }))
    return pd.concat(df_list, ignore_index=True)


# ================================================================
# 📘 ФУНКЦИЯ RM-ANOVA
# ================================================================
def rm_anova(data_dict):
    """
    data_dict — словарь: имя_условия → список значений

    Возвращает:
        таблицу ANOVA (F, p-value)
    """

    df = make_rm_dataframe(data_dict)

    model = AnovaRM(
        df,
        depvar="value",
        subject="id",
        within=["condition"]
    )

    return model.fit()


# ================================================================
# 📘 5.1.1 Пример — скорость работы утром/днём/вечером
# ================================================================
morning = [10, 12, 11, 9, 14]
day     = [12, 15, 13, 11, 16]
evening = [14, 16, 15, 13, 18]

res1 = rm_anova({
    "morning": morning,
    "day": day,
    "evening": evening
})
print("5.1.1 morning/day/evening:")
print(res1)


# ================================================================
# 📘 5.1.2 Увеличение производительности по времени (t1 / t2 / t3)
# ================================================================
t1 = [100, 110, 120, 115, 108]
t2 = [130, 125, 140, 145, 135]
t3 = [150, 160, 155, 165, 158]

res2 = rm_anova({"t1": t1, "t2": t2, "t3": t3})
print("\n5.1.2 t1/t2/t3:")
print(res2)


# ================================================================
# 📘 5.1.3 Тестирование интерфейсов A/B/C на тех же сотрудниках
# ================================================================
A = [20, 22, 25, 19, 21]
B = [18, 17, 19, 20, 18]
C = [16, 14, 15, 17, 16]

res3 = rm_anova({"A": A, "B": B, "C": C})
print("\n5.1.3 interface A/B/C:")
print(res3)


# ================================================================
# 📘 5.1.4 Качество сна за 3 недели у тех же людей
# ================================================================
week1 = [7, 6, 8, 7, 5]
week2 = [6, 7, 7, 8, 6]
week3 = [8, 8, 9, 7, 7]

res4 = rm_anova({"w1": week1, "w2": week2, "w3": week3})
print("\n5.1.4 sleep quality:")
print(res4)


# ================================================================
# 📘 5.1.5 Нормальные данные с выбросом (как тест устойчивости)
# ================================================================
c1 = [10, 11, 12, 10, 50]   # выброс
c2 = [9, 10, 11, 9, 12]
c3 = [11, 12, 13, 11, 14]

res5 = rm_anova({"c1": c1, "c2": c2, "c3": c3})
print("\n5.1.5 outlier check:")
print(res5)


# ================================================================
# 📘 5.1.6 Случайные данные: равные средние → p должно быть большим
# ================================================================
g1 = np.random.normal(50, 5, 50)
g2 = np.random.normal(50, 5, 50)
g3 = np.random.normal(50, 5, 50)

res6 = rm_anova({"g1": g1, "g2": g2, "g3": g3})
print("\n5.1.6 equal means (should be non-sig):")
print(res6)


# ================================================================
# 📘 5.1.7 Сильные различия между условиями → p очень маленькое
# ================================================================
g1 = np.random.normal(10, 2, 40)
g2 = np.random.normal(20, 2, 40)
g3 = np.random.normal(30, 2, 40)

res7 = rm_anova({"g1": g1, "g2": g2, "g3": g3})
print("\n5.1.7 strong effect:")
print(res7)


# ================================================================
# 📘 5.1.8 Перекос распределений (RM-ANOVA не идеален, но работает)
# ================================================================
exp1 = np.random.exponential(1, 50)
exp2 = np.random.exponential(1.5, 50)
exp3 = np.random.exponential(2, 50)

res8 = rm_anova({"e1": exp1, "e2": exp2, "e3": exp3})
print("\n5.1.8 exponential data:")
print(res8)


# ================================================================
# 📘 5.1.9 Бизнес-кейс — производительность по сменам (shift1/2/3)
# ================================================================
shift1 = np.random.normal(100, 20, 200)
shift2 = np.random.normal(120, 25, 200)
shift3 = np.random.normal(140, 30, 200)

res9 = rm_anova({"shift1": shift1, "shift2": shift2, "shift3": shift3})
print("\n5.1.9 shifts performance:")
print(res9)


# ================================================================
# 📘 Итог 5.1 RM-ANOVA
# ------------------------------------------------
# ✔ подходит для 3+ зависимых условий
# ✔ нормальность и сферичность — важные предпосылки
# ✔ мощнее Фридмана, если условия соблюдены
#
# Альтернативы:
#   • 5.2 Friedman test — если данные ненормальные
#   • 5.3 Repeated-measures permutation — если хотим избежать предпосылок
# ------------------------------------------------
# ================================================================
```

```python
# ================================================================
# 📘 5.1.a Linear Mixed Models (LMM)
# ================================================================
# Когда применять:
#   ✔ 3+ зависимых измерений (утро/день/вечер и т.п.)
#   ✔ пропуски есть → RM-ANOVA НЕ работает
#   ✔ у разных сотрудников разное количество наблюдений
#   ✔ интервалы между измерениями НЕ равные
#   ✔ дисперсии разные
#   ✔ нужны индивидуальные случайные эффекты (id)
#
# Проверяет:
#   H0: нет эффекта фактора (например, нет разницы между утро/день/вечер)
#
# Альтернативы:
#   • 5.1 RM-ANOVA — требует полную таблицу без пропусков
#   • 5.2 Friedman — ранговый аналог, без индивидуальных наклонов
#   • LMM — самый гибкий метод
# ================================================================

import numpy as np
import pandas as pd
from statsmodels.formula.api import mixedlm


# ================================================================
# 📘 ВСПОМОГАТЕЛЬНАЯ ФУНКЦИЯ — правильное формирование таблицы
# ================================================================
def make_lmm_dataframe(data_dict):
    """
    Преобразует словарь формата:
        {"morning":[...], "day":[...], "evening":[...]}
    в таблицу формата long:

         id | condition | value

    Ключевой момент — выравнивание количества ID по max длине.
    """
    df_parts = []

    max_len = max(len(v) for v in data_dict.values())

    for condition, values in data_dict.items():
        # увеличиваем список до max_len, если он короче
        values = list(values) + [np.nan] * (max_len - len(values))

        df_parts.append(pd.DataFrame({
            "id": np.arange(max_len),
            "condition": condition,
            "value": values
        }))

    df = pd.concat(df_parts, ignore_index=True)
    return df


# ================================================================
# 📘 ОСНОВНАЯ ФУНКЦИЯ LMM
# ================================================================
def lmm_test(data_dict):
    """
    data_dict: словарь вида
        { "t1": [...], "t2":[...], ... }

    Возвращает:
        модель MixedLM + результат summary()
    """

    df = make_lmm_dataframe(data_dict)

    # удаляем строки, где нет значения
    df = df.dropna(subset=["value"])

    model = mixedlm(
        "value ~ condition",   # фиксированный эффект
        df,
        groups=df["id"]        # случайный перехват по сотруднику
    )

    return model.fit()


# ================================================================
# 📘 5.1.a.1 — Пропуски в измерениях
# ================================================================
morning = [10, 12, 11,  9, 14]
day     = [np.nan, 15, 13, 11, 16]
evening = [14, 16, np.nan, 13, 18]

res1 = lmm_test({
    "morning": morning,
    "day": day,
    "evening": evening
})

print("\n==================== 5.1.a.1 Missing values ====================")
print(res1.summary())


# ================================================================
# 📘 5.1.a.2 — Нерегулярные интервалы (месяцы 1,2,6)
# ================================================================
month1 = [100, 110,  np.nan, 115]
month2 = [130, 125, 140,      np.nan]
month6 = [150, 160, 155,      np.nan]

res2 = lmm_test({"m1": month1, "m2": month2, "m6": month6})

print("\n==================== 5.1.a.2 Irregular intervals ====================")
print(res2.summary())


# ================================================================
# 📘 5.1.a.3 — Разное количество измерений у сотрудников
# ================================================================
c1 = [20, 22, 25, 19]            # 4 измерения
c2 = [18, np.nan, 19]            # 3 измерения
c3 = [16, 14]                    # 2 измерения

res3 = lmm_test({"c1": c1, "c2": c2, "c3": c3})

print("\n==================== 5.1.a.3 Unbalanced ====================")
print(res3.summary())


# ================================================================
# 📘 5.1.a.4 — Индивидуальные траектории (рост)
# ================================================================
p1 = [1, 2, 3, 4, 5]
p2 = [1, 2, np.nan, 4, 5]
p3 = [1, np.nan, 3, 4, 5]

res4 = lmm_test({"t1": p1, "t2": p2, "t3": p3})

print("\n==================== 5.1.a.4 Growth curves ====================")
print(res4.summary())


# ================================================================
# 📘 5.1.a.5 — Большая разница дисперсий
# ================================================================
g1 = [10, 11, 12, 10, 9]
g2 = [50, 70, 65, 90, 80]               # огромная дисперсия
g3 = [15, 16, 14, 17, 16]

res5 = lmm_test({"g1": g1, "g2": g2, "g3": g3})

print("\n==================== 5.1.a.5 Heteroscedasticity ====================")
print(res5.summary())


# ================================================================
# 📘 5.1.a.6 — Реальный бизнес-кейс: производительность по месяцам
# ================================================================
jan = np.random.normal(100, 20, 200)
feb = np.random.normal(120, 25, 200)
mar = np.random.normal(140, 35, 200)

jan[5] = np.nan
feb[10] = np.nan

res6 = lmm_test({"jan": jan, "feb": feb, "mar": mar})

print("\n==================== 5.1.a.6 Business case ====================")
print(res6.summary())


# ================================================================
# 📘 ИТОГ
# ------------------------------------------------
# ✔ Поддерживает пропуски
# ✔ Поддерживает разное число измерений
# ✔ Поддерживает разные дисперсии
# ✔ Поддерживает нерегулярные интервалы
# ✔ Моделирует индивидуальные различия (random effects)
# ------------------------------------------------
# MixedLM — самый мощный метод для повторных измерений
# ================================================================
```
